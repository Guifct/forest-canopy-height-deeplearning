```python
# ============================================================
# 2. DATASET E AUGMENTATION
# ============================================================

class FCHDataset(Dataset):
    def __init__(self, X, y, transform=None):
        self.X = X.astype(np.float32)
        self.y = y.astype(np.float32)
        self.transform = transform

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        x = self.X[idx].copy()
        if self.transform is not None:
            x = self.transform(x)
        return (
            torch.tensor(x, dtype=torch.float32),
            torch.tensor(self.y[idx], dtype=torch.float32),
        )


class NormalizedIndexedFCHDataset(Dataset):
    """
    Dataset memory-safe.

    Em vez de criar X_train e X_val normalizados como arrays enormes,
    guarda apenas X_all + índices. Cada patch é normalizado no __getitem__.

    Isto evita erros do tipo:
    MemoryError: Unable to allocate ... bool array with shape (N, C, H, W)
    """
    def __init__(self, X_all, y_all, indices, mean, std, transform=None, target_scale=1.0):
        self.X_all = X_all
        self.y_all = y_all
        self.indices = np.asarray(indices, dtype=np.int64)
        self.mean = np.asarray(mean, dtype=np.float32).reshape(-1, 1, 1)
        self.std = np.asarray(std, dtype=np.float32).reshape(-1, 1, 1)
        self.transform = transform
        self.target_scale = float(target_scale)

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        raw_idx = self.indices[idx]

        # Copia só uma amostra. Não copia o fold inteiro.
        x = np.array(self.X_all[raw_idx], dtype=np.float32, copy=True)

        # Normalização in-place.
        np.subtract(x, self.mean, out=x)
        np.divide(x, self.std, out=x)
        np.nan_to_num(x, copy=False, nan=0.0, posinf=0.0, neginf=0.0)

        if self.transform is not None:
            x = self.transform(x)

        x = np.ascontiguousarray(x, dtype=np.float32)
        y = np.float32(self.y_all[raw_idx] / self.target_scale)

        return torch.from_numpy(x), torch.tensor(y, dtype=torch.float32)


def base_augmentation(x):
    """Rotações e flips aleatórios. Usar apenas no treino."""
    k = np.random.randint(0, 4)
    if k > 0:
        x = np.rot90(x, k, axes=(1, 2)).copy()

    if np.random.rand() > 0.5:
        x = np.flip(x, axis=2).copy()

    if np.random.rand() > 0.5:
        x = np.flip(x, axis=1).copy()

    return x


def get_train_transform(augmentation_mode):
    """Devolve a transformação de treino para a experiência de ablation."""
    if augmentation_mode == "aug_on":
        return base_augmentation

    if augmentation_mode == "aug_off":
        return None

    raise ValueError(
        f"augmentation_mode inválido: {augmentation_mode}. "
        "Usa 'aug_on' ou 'aug_off'."
    )
```

```python
# ============================================================
# 3. CARREGAMENTO DOS TIFs
# ============================================================



def chunked_nan_percentage(X, chunk_size=128):
    """
    Calcula a percentagem de NaNs em X sem criar uma máscara booleana gigante.

    Evita operações do tipo np.isnan(X) sobre o array inteiro, que podem duplicar
    centenas de MB em memória. Processa apenas pequenos blocos de amostras.
    """
    n = int(X.shape[0])
    total_values = int(np.prod(X.shape))
    if total_values == 0:
        return 0.0

    nan_count = 0
    for start in range(0, n, int(chunk_size)):
        end = min(start + int(chunk_size), n)
        block = X[start:end]
        nan_count += int(np.isnan(block).sum())

    return 100.0 * nan_count / total_values


def extract_point_id(filename):
    m = re.search(r"point_(\d+)_", filename)
    if m is None:
        raise ValueError(f"Não consegui extrair point_id de: {filename}")
    return int(m.group(1))


def build_target_dict(gdf, target_column):
    possible_id_cols = ["point_id", "POINT_ID", "id", "ID", "fid", "FID"]
    id_col = next((c for c in possible_id_cols if c in gdf.columns), None)

    ids = gdf[id_col].astype(int).values if id_col else gdf.index.astype(int).values
    targets = gdf[target_column].values

    return {
        int(pid): float(t)
        for pid, t in zip(ids, targets)
        if not pd.isna(t)
    }


def canonical_band_name(name):
    """
    Normaliza nomes de bandas para comparar regiões.

    Remove:
      - prefixo da área, se existir antes de 20200000;
      - sufixos técnicos que não mudam a variável, como _A e _Mosaic.

    Exemplo:
      SerraLousa_20200000_S1SLC_COHVH_Mosaic -> 20200000_S1SLC_COHVH
      Mafra_20200000_S1SLC_COHVH_A           -> 20200000_S1SLC_COHVH
    """
    s = str(name)

    # Fica só com a parte a partir do ano, se existir.
    m = re.search(r"20\d{6}_.+", s)
    if m:
        s = m.group(0)

    # Remove sufixos finais que representam versão/produto, não variável física.
    s = re.sub(r"_(A|MOSAIC)$", "", s, flags=re.IGNORECASE)

    return s


def get_valid_band_indices(metadata_path):
    with open(metadata_path, "r", encoding="utf-8") as f:
        meta = json.load(f)

    raster_names = meta.get("raster_names", [])

    valid_indices = [
        i for i, name in enumerate(raster_names)
        if "CHM" not in name.upper()
        and "LIDAR" not in name.upper()
        and "OTHER" not in name.upper()
    ]

    removed = [
        name for name in raster_names
        if "CHM" in name.upper()
        or "LIDAR" in name.upper()
        or "OTHER" in name.upper()
    ]

    if removed:
        print(f"   Bandas removidas (leakage): {removed}")

    return valid_indices, raster_names


def find_common_region_point_ids(patch_base_dir, patch_sizes, validation_dir, target_column="rh_opt_098"):
    """
    Encontra os pontos (região, point_id) presentes em todos os tamanhos de patch.
    Não lê os rasters para memória; apenas varre os nomes dos GeoTIFFs.
    """
    patch_base_dir = Path(patch_base_dir)
    validation_dir = Path(validation_dir)

    ids_by_size = {}

    for patch_size in patch_sizes:
        raw_base_dir = patch_base_dir / f"patch_{patch_size}x{patch_size}"
        ids_this_size = set()

        if not raw_base_dir.exists():
            raise FileNotFoundError(f"Pasta não encontrada: {raw_base_dir}")

        for area_dir in sorted(raw_base_dir.iterdir()):
            if not area_dir.is_dir():
                continue

            area_name = area_dir.name
            gpkg_path = validation_dir / f"{area_name}_HEIGHT_Dataset.gpkg"
            if not gpkg_path.exists():
                continue

            gdf = gpd.read_file(gpkg_path)
            target_dict = build_target_dict(gdf, target_column)

            for tif in area_dir.rglob("*.tif"):
                pid = extract_point_id(tif.name)
                if pid in target_dict:
                    ids_this_size.add((area_name, int(pid)))

        ids_by_size[patch_size] = ids_this_size
        print(f"Patch {patch_size}x{patch_size}: {len(ids_this_size)} pontos com target")

    common_ids = set.intersection(*ids_by_size.values())
    print(f"\nPontos comuns a todos os patch sizes: {len(common_ids)}")

    return common_ids, ids_by_size


def load_all_regions_tifs(
    raw_base_dir,
    validation_dir,
    target_column="rh_opt_098",
    allowed_region_point_ids=None,
):
    raw_base_dir = Path(raw_base_dir)
    validation_dir = Path(validation_dir)

    if not raw_base_dir.exists():
        raise FileNotFoundError(f"Pasta de patches não encontrada: {raw_base_dir}")

    X_list, y_list, region_list, pid_list = [], [], [], []
    channels_ref = None
    band_names_ref = None
    canonical_band_names_ref = None

    print("-> A ler TIFs e cruzar com GeoPackages...")
    print(f"   Pasta: {raw_base_dir}")

    for area_dir in sorted(raw_base_dir.iterdir()):
        if not area_dir.is_dir():
            continue

        area_name = area_dir.name
        gpkg_path = validation_dir / f"{area_name}_HEIGHT_Dataset.gpkg"

        if not gpkg_path.exists():
            print(f"   [{area_name}] GeoPackage não encontrado — ignorado.")
            continue

        metas = list(area_dir.glob("*.json"))
        if not metas:
            print(f"   [{area_name}] metadata JSON não encontrado — ignorado.")
            continue

        print(f"\nÁrea: {area_name}")

        gdf = gpd.read_file(gpkg_path)
        target_dict = build_target_dict(gdf, target_column)
        valid_idx, raster_names = get_valid_band_indices(metas[0])
        valid_band_names = [raster_names[i] for i in valid_idx]
        canonical_band_names = [canonical_band_name(name) for name in valid_band_names]

        if channels_ref is None:
            channels_ref = len(valid_idx)
            band_names_ref = valid_band_names
            canonical_band_names_ref = canonical_band_names
            print("   Área de referência para ordem das bandas.")
        else:
            if len(valid_idx) != channels_ref:
                raise ValueError(
                    f"Canais incompatíveis em {area_name}: "
                    f"esperado {channels_ref}, obtido {len(valid_idx)}"
                )

            same_order = canonical_band_names == canonical_band_names_ref
            same_set = set(canonical_band_names) == set(canonical_band_names_ref)

            if not same_order:
                if not same_set:
                    missing = sorted(set(canonical_band_names_ref) - set(canonical_band_names))
                    extra = sorted(set(canonical_band_names) - set(canonical_band_names_ref))
                    raise ValueError(
                        f"Bandas incompatíveis em {area_name}.\n"
                        f"Faltam: {missing[:20]}\n"
                        f"Extra: {extra[:20]}"
                    )

                # Se tiver as mesmas bandas mas em ordem diferente, reordena.
                key_to_raster_index = {
                    canonical_band_name(name): i
                    for i, name in enumerate(raster_names)
                }
                valid_idx = [key_to_raster_index[key] for key in canonical_band_names_ref]
                print(f"   AVISO: bandas reordenadas em {area_name} para seguir a referência.")
            else:
                print("   Ordem das bandas OK.")

        n_loaded = 0
        n_no_target = 0
        n_filtered_common = 0

        for tif in sorted(area_dir.rglob("*.tif")):
            pid = extract_point_id(tif.name)

            if pid not in target_dict:
                n_no_target += 1
                continue

            if allowed_region_point_ids is not None and (area_name, int(pid)) not in allowed_region_point_ids:
                n_filtered_common += 1
                continue

            with rasterio.open(tif) as src:
                patch = src.read().astype(np.float32)[valid_idx]

            X_list.append(patch)
            y_list.append(target_dict[pid])
            region_list.append(area_name)
            pid_list.append(pid)
            n_loaded += 1

        msg = f"   Carregados: {n_loaded} | Sem target: {n_no_target}"
        if allowed_region_point_ids is not None:
            msg += f" | Fora da interseção comum: {n_filtered_common}"
        print(msg)

    if not X_list:
        raise RuntimeError("Nenhum patch carregado.")

    X = np.stack(X_list).astype(np.float32)
    y = np.array(y_list, dtype=np.float32)
    regions = np.array(region_list)
    point_ids = np.array(pid_list)

    print(f"\nResumo: X={X.shape} | y={y.shape} | regiões={sorted(set(regions))}")
    print(f"NaNs em X: {chunked_nan_percentage(X, chunk_size=NORMALIZATION_CHUNK_SIZE):.6f}%")

    return X, y, regions, point_ids, band_names_ref
```

```python
# ============================================================
# 4. NORMALIZAÇÃO SEM LEAKAGE E SEM CÓPIAS GIGANTES
# ============================================================

NORMALIZATION_STATS_CACHE = {}


def compute_channel_stats_from_indices(X_all, indices, chunk_size=NORMALIZATION_CHUNK_SIZE):
    """
    Calcula média e desvio padrão por canal usando apenas o treino do fold.

    Versão segura em memória:
    - não faz X_all[train_idx] para o fold inteiro;
    - processa chunks pequenos;
    - cria máscaras booleanas só para um chunk de cada vez.
    """
    indices = np.asarray(indices, dtype=np.int64)
    n_channels = X_all.shape[1]

    sum_x = np.zeros(n_channels, dtype=np.float64)
    sum_x2 = np.zeros(n_channels, dtype=np.float64)
    count = np.zeros(n_channels, dtype=np.float64)

    for start in range(0, len(indices), chunk_size):
        batch_idx = indices[start:start + chunk_size]

        # Fancy indexing copia só este chunk, não o fold inteiro.
        xb = np.asarray(X_all[batch_idx], dtype=np.float32)

        finite = np.isfinite(xb)
        count += finite.sum(axis=(0, 2, 3), dtype=np.float64)

        # Zera NaN/inf para não contaminarem soma e soma dos quadrados.
        np.nan_to_num(xb, copy=False, nan=0.0, posinf=0.0, neginf=0.0)

        sum_x += xb.sum(axis=(0, 2, 3), dtype=np.float64)
        sum_x2 += np.einsum("nchw,nchw->c", xb, xb, dtype=np.float64)

        del xb, finite

    count = np.maximum(count, 1.0)
    mean = sum_x / count
    var = (sum_x2 / count) - (mean ** 2)
    var = np.maximum(var, 1e-12)
    std = np.sqrt(var)

    std[std < 1e-6] = 1.0

    return mean.astype(np.float32).reshape(-1, 1, 1), std.astype(np.float32).reshape(-1, 1, 1)


def get_cached_channel_stats(X_all, train_idx):
    """Reutiliza estatísticas do mesmo fold entre arquiteturas."""
    train_idx = np.asarray(train_idx, dtype=np.int64)
    cache_key = tuple(train_idx.tolist())

    if cache_key not in NORMALIZATION_STATS_CACHE:
        print("A calcular mean/std do fold em modo memory-safe...")
        NORMALIZATION_STATS_CACHE[cache_key] = compute_channel_stats_from_indices(
            X_all,
            train_idx,
            chunk_size=NORMALIZATION_CHUNK_SIZE,
        )
    else:
        print("A reutilizar mean/std do fold a partir da cache.")

    return NORMALIZATION_STATS_CACHE[cache_key]


def normalize_from_train(X_train_raw, X_val_raw):
    """
    Função antiga mantida só para compatibilidade.

    Para este notebook, evita usar isto em folds grandes, porque cria arrays
    normalizados completos em memória. O treino agora usa NormalizedIndexedFCHDataset.
    """
    mean = np.nanmean(X_train_raw, axis=(0, 2, 3), keepdims=True).astype(np.float32)
    std = np.nanstd(X_train_raw, axis=(0, 2, 3), keepdims=True).astype(np.float32)
    std[std < 1e-6] = 1.0

    X_train = ((X_train_raw - mean) / std).astype(np.float32)
    X_val = ((X_val_raw - mean) / std).astype(np.float32)
    np.nan_to_num(X_train, copy=False, nan=0.0, posinf=0.0, neginf=0.0)
    np.nan_to_num(X_val, copy=False, nan=0.0, posinf=0.0, neginf=0.0)

    return X_train, X_val, mean, std
```

```python
# ============================================================
# 5. ARQUITETURAS MODULARES — FUSION + BLOCKS + POOLING + REGRESSOR
# ============================================================

from collections import OrderedDict


# ------------------------------------------------------------
# Agrupamento dos canais por sensor
# ------------------------------------------------------------

def infer_sensor_group_from_name(name):
    """Heurística para agrupar canais por tipo de sensor/variável."""
    u = str(name).upper()

    # Sentinel-1 / SAR
    if any(k in u for k in ["SENTINEL1", "SENTINEL-1", "S1", "_VV", "_VH", "VV_", "VH_", "SAR"]):
        return "S1_SAR"

    # Sentinel-2 / óptico/multiespectral
    s2_band_tokens = [f"B{i:02d}" for i in range(1, 13)] + ["B8A", "NDVI", "EVI", "NDWI", "NDMI", "NBR"]
    if any(k in u for k in ["SENTINEL2", "SENTINEL-2", "S2"] + s2_band_tokens):
        return "S2_OPTICAL"

    # Topografia ou variáveis auxiliares não derivadas diretamente do target
    if any(k in u for k in ["DEM", "DTM", "DSM", "ELEV", "ELEVATION", "SLOPE", "ASPECT", "TPI", "TRI"]):
        return "TOPOGRAPHY"

    # Máscaras/indicadores de disponibilidade
    if any(k in u for k in ["MASK", "VALID", "AVAIL", "NODATA"]):
        return "MASKS"

    return "OTHER"


def infer_late_fusion_channel_groups(band_names):
    """
    Cria grupos de canais para late/hybrid fusion.

    Cada grupo pode ser processado por uma branch separada. Isto permite comparar:
      - early fusion: todos os canais juntos desde o início;
      - late fusion: cada grupo/sensor tem branch própria até à fusão final;
      - hybrid fusion: cada grupo tem um stem próprio, mas os mapas são fundidos antes dos blocos comuns.
    """
    groups = OrderedDict()

    for idx, name in enumerate(band_names):
        group = infer_sensor_group_from_name(name)
        groups.setdefault(group, []).append(idx)

    preferred_order = ["S1_SAR", "S2_OPTICAL", "TOPOGRAPHY", "MASKS", "OTHER"]
    ordered = OrderedDict()
    for group in preferred_order:
        if group in groups:
            ordered[group] = groups[group]
    for group in groups:
        if group not in ordered:
            ordered[group] = groups[group]

    print("\nGrupos de canais detetados:")
    for group, idxs in ordered.items():
        print(f"  {group}: {len(idxs)} canais")

    if len(ordered) < 2:
        print(
            "AVISO: só foi detetado um grupo de canais. "
            "Early, late e hybrid fusion ficam quase equivalentes."
        )

    return ordered


# ------------------------------------------------------------
# Blocos CNN
# ------------------------------------------------------------

class ConvBlock(nn.Module):
    """Bloco base: Conv2d -> BatchNorm -> GELU."""
    def __init__(self, in_channels, out_channels, kernel_size=3, dilation=1):
        super().__init__()
        padding = dilation * (kernel_size // 2)
        self.block = nn.Sequential(
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=kernel_size,
                padding=padding,
                dilation=dilation,
                bias=False,
            ),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
        )

    def forward(self, x):
        return self.block(x)


class DoubleConvBlock(nn.Module):
    """Dois blocos convolucionais seguidos. Aumenta capacidade local."""
    def __init__(self, in_channels, out_channels, kernel_size=3, dilation=1):
        super().__init__()
        padding = dilation * (kernel_size // 2)
        self.block = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size, padding=padding, dilation=dilation, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
            nn.Conv2d(out_channels, out_channels, kernel_size, padding=padding, dilation=dilation, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
        )

    def forward(self, x):
        return self.block(x)


class SEBlock(nn.Module):
    """Squeeze-and-Excitation: recalibra canais/features."""
    def __init__(self, channels, reduction=8):
        super().__init__()
        hidden = max(channels // reduction, 4)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, hidden),
            nn.GELU(),
            nn.Linear(hidden, channels),
            nn.Sigmoid(),
        )

    def forward(self, x):
        b, c, _, _ = x.shape
        weights = self.pool(x).view(b, c)
        weights = self.fc(weights).view(b, c, 1, 1)
        return x * weights


class SEConvBlock(nn.Module):
    """ConvBlock + SE attention."""
    def __init__(self, in_channels, out_channels, kernel_size=3, dilation=1, reduction=8):
        super().__init__()
        self.conv = ConvBlock(in_channels, out_channels, kernel_size=kernel_size, dilation=dilation)
        self.se = SEBlock(out_channels, reduction=reduction)

    def forward(self, x):
        return self.se(self.conv(x))


class ResidualBlock(nn.Module):
    """Bloco residual: aprende uma correção em cima da entrada."""
    def __init__(self, in_channels, out_channels, kernel_size=3, dilation=1):
        super().__init__()
        padding = dilation * (kernel_size // 2)

        self.conv1 = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size, padding=padding, dilation=dilation, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.GELU(),
        )
        self.conv2 = nn.Sequential(
            nn.Conv2d(out_channels, out_channels, kernel_size, padding=padding, dilation=dilation, bias=False),
            nn.BatchNorm2d(out_channels),
        )
        self.proj = None
        if in_channels != out_channels:
            self.proj = nn.Sequential(
                nn.Conv2d(in_channels, out_channels, kernel_size=1, bias=False),
                nn.BatchNorm2d(out_channels),
            )
        self.act = nn.GELU()

    def forward(self, x):
        identity = x if self.proj is None else self.proj(x)
        out = self.conv1(x)
        out = self.conv2(out)
        return self.act(out + identity)


class ResidualSEBlock(nn.Module):
    """ResidualBlock + SE attention."""
    def __init__(self, in_channels, out_channels, kernel_size=3, dilation=1, reduction=8):
        super().__init__()
        self.res = ResidualBlock(in_channels, out_channels, kernel_size=kernel_size, dilation=dilation)
        self.se = SEBlock(out_channels, reduction=reduction)

    def forward(self, x):
        return self.se(self.res(x))


def make_spatial_block(block_type, in_channels, out_channels, dilation=1, se_reduction=8):
    """Constrói um bloco CNN a partir de uma string."""
    block_type = str(block_type).lower()

    if block_type == "conv":
        return ConvBlock(in_channels, out_channels, kernel_size=3, dilation=dilation)
    if block_type == "double_conv":
        return DoubleConvBlock(in_channels, out_channels, kernel_size=3, dilation=dilation)
    if block_type == "residual":
        return ResidualBlock(in_channels, out_channels, kernel_size=3, dilation=dilation)
    if block_type == "se":
        return SEConvBlock(in_channels, out_channels, kernel_size=3, dilation=dilation, reduction=se_reduction)
    if block_type == "residual_se":
        return ResidualSEBlock(in_channels, out_channels, kernel_size=3, dilation=dilation, reduction=se_reduction)
    if block_type == "dilated":
        # Mantém bloco simples, mas força convoluções dilatadas.
        return ConvBlock(in_channels, out_channels, kernel_size=3, dilation=max(int(dilation), 2))

    raise ValueError(
        f"block_type inválido: {block_type}. "
        "Usa: conv, double_conv, residual, se, residual_se ou dilated."
    )


def make_dilation_schedule(depth, dilation_mode="last"):
    """
    Define a dilatação dos blocos.

    none        -> todos os blocos com dilation=1
    last        -> último bloco com dilation=2
    alternating -> alterna dilation 1 e 2
    all2        -> todos os blocos com dilation=2
    """
    depth = int(depth)
    if depth <= 0:
        return []

    dilation_mode = str(dilation_mode).lower()
    if dilation_mode == "none":
        return [1] * depth
    if dilation_mode == "last":
        return [1] * (depth - 1) + [2]
    if dilation_mode == "alternating":
        return [1 if i % 2 == 0 else 2 for i in range(depth)]
    if dilation_mode == "all2":
        return [2] * depth

    raise ValueError("dilation_mode inválido. Usa: none, last, alternating ou all2.")


# ------------------------------------------------------------
# Pooling espacial
# ------------------------------------------------------------

class SpatialPooler(nn.Module):
    """
    Converte mapas de features [B, C, H, W] num vetor.

    pooling_type:
      - gap1: Global Average Pooling 1x1
      - gap2: Global Average Pooling 2x2
      - gap3: Global Average Pooling 3x3
      - center_only: só janela central
      - gap2_center: GAP 2x2 + janela central
      - gap2_center_max: GAP 2x2 + center + Global Max Pooling 2x2
    """
    def __init__(self, channels, pooling_type="gap2_center", center_radius=1):
        super().__init__()
        self.channels = int(channels)
        self.pooling_type = str(pooling_type).lower()
        self.center_radius = int(center_radius)

        self.use_avg = False
        self.use_max = False
        self.use_center = False
        self.global_pool_size = 1

        if self.pooling_type == "gap1":
            self.use_avg = True
            self.global_pool_size = 1
        elif self.pooling_type == "gap2":
            self.use_avg = True
            self.global_pool_size = 2
        elif self.pooling_type == "gap3":
            self.use_avg = True
            self.global_pool_size = 3
        elif self.pooling_type == "center_only":
            self.use_center = True
        elif self.pooling_type == "gap2_center":
            self.use_avg = True
            self.use_center = True
            self.global_pool_size = 2
        elif self.pooling_type == "gap2_center_max":
            self.use_avg = True
            self.use_center = True
            self.use_max = True
            self.global_pool_size = 2
        else:
            raise ValueError(
                f"pooling_type inválido: {pooling_type}. "
                "Usa: gap1, gap2, gap3, center_only, gap2_center, gap2_center_max."
            )

        if self.use_avg:
            self.avg_pool = nn.AdaptiveAvgPool2d((self.global_pool_size, self.global_pool_size))
        if self.use_max:
            self.max_pool = nn.AdaptiveMaxPool2d((self.global_pool_size, self.global_pool_size))

        out_features = 0
        if self.use_avg:
            out_features += self.channels * self.global_pool_size * self.global_pool_size
        if self.use_center:
            out_features += self.channels
        if self.use_max:
            out_features += self.channels * self.global_pool_size * self.global_pool_size
        self.out_features = int(out_features)

    def extract_center_features(self, x):
        _, _, h, w = x.shape
        cy, cx = h // 2, w // 2
        y0 = max(cy - self.center_radius, 0)
        y1 = min(cy + self.center_radius + 1, h)
        x0 = max(cx - self.center_radius, 0)
        x1 = min(cx + self.center_radius + 1, w)
        return x[:, :, y0:y1, x0:x1].mean(dim=(2, 3))

    def forward(self, x):
        features = []
        if self.use_avg:
            features.append(torch.flatten(self.avg_pool(x), start_dim=1))
        if self.use_center:
            features.append(self.extract_center_features(x))
        if self.use_max:
            features.append(torch.flatten(self.max_pool(x), start_dim=1))
        return torch.cat(features, dim=1)


class ArchitectureSpatialBranch(nn.Module):
    """Branch CNN configurável para early ou late fusion."""
    def __init__(
        self,
        in_channels,
        branch_channels=32,
        depth=3,
        block_type="conv",
        pooling_type="gap2_center",
        center_radius=1,
        dilation_mode="last",
        se_reduction=8,
    ):
        super().__init__()
        self.in_channels = int(in_channels)
        self.branch_channels = int(branch_channels)
        self.depth = int(depth)
        self.block_type = str(block_type).lower()
        self.pooling_type = str(pooling_type).lower()

        # Stem 1x1: mistura canais do sensor/grupo sem misturar espaço.
        self.stem = ConvBlock(self.in_channels, self.branch_channels, kernel_size=1, dilation=1)

        dilations = make_dilation_schedule(self.depth, dilation_mode=dilation_mode)
        blocks = []
        for d in dilations:
            blocks.append(
                make_spatial_block(
                    self.block_type,
                    self.branch_channels,
                    self.branch_channels,
                    dilation=d,
                    se_reduction=se_reduction,
                )
            )
        self.blocks = nn.Sequential(*blocks)
        self.pooler = SpatialPooler(
            channels=self.branch_channels,
            pooling_type=self.pooling_type,
            center_radius=center_radius,
        )
        self.out_features = self.pooler.out_features

    def forward(self, x):
        x = self.stem(x)
        x = self.blocks(x)
        return self.pooler(x)


# ------------------------------------------------------------
# Regressor final
# ------------------------------------------------------------

class ResidualMLPBlock(nn.Module):
    def __init__(self, dim, dropout=0.2):
        super().__init__()
        self.block = nn.Sequential(
            nn.Linear(dim, dim, bias=False),
            nn.BatchNorm1d(dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(dim, dim, bias=False),
            nn.BatchNorm1d(dim),
        )
        self.act = nn.GELU()

    def forward(self, x):
        return self.act(x + self.block(x))


def make_regressor(in_features, regressor_type="current_mlp", dropout=0.3):
    """Constrói o MLP final."""
    regressor_type = str(regressor_type).lower()
    in_features = int(in_features)

    if regressor_type == "linear":
        return nn.Linear(in_features, 1)

    if regressor_type == "small_mlp":
        return nn.Sequential(
            nn.Linear(in_features, 64, bias=False),
            nn.BatchNorm1d(64),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(64, 1),
        )

    if regressor_type == "current_mlp":
        return nn.Sequential(
            nn.Linear(in_features, 128, bias=False),
            nn.BatchNorm1d(128),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(128, 64),
            nn.GELU(),
            nn.Dropout(max(dropout - 0.1, 0.0)),
            nn.Linear(64, 1),
        )

    if regressor_type == "deep_mlp":
        return nn.Sequential(
            nn.Linear(in_features, 256, bias=False),
            nn.BatchNorm1d(256),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(256, 128, bias=False),
            nn.BatchNorm1d(128),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(128, 64),
            nn.GELU(),
            nn.Dropout(max(dropout - 0.1, 0.0)),
            nn.Linear(64, 1),
        )

    if regressor_type == "residual_mlp":
        hidden = 128
        return nn.Sequential(
            nn.Linear(in_features, hidden, bias=False),
            nn.BatchNorm1d(hidden),
            nn.GELU(),
            ResidualMLPBlock(hidden, dropout=dropout),
            nn.Dropout(dropout),
            nn.Linear(hidden, 64),
            nn.GELU(),
            nn.Linear(64, 1),
        )

    raise ValueError(
        f"regressor_type inválido: {regressor_type}. "
        "Usa: linear, small_mlp, current_mlp, deep_mlp ou residual_mlp."
    )


# ------------------------------------------------------------
# Modelos de fusão
# ------------------------------------------------------------

def _sanitize_module_key(name):
    return re.sub(r"[^A-Za-z0-9_]", "_", str(name))


def _validate_channel_groups(in_channels, channel_groups):
    group_names = list(channel_groups.keys())
    group_indices = [list(channel_groups[g]) for g in group_names]

    used_channels = sorted([idx for idxs in group_indices for idx in idxs])
    if len(used_channels) == 0:
        raise ValueError("Nenhum canal foi selecionado para a arquitetura.")

    invalid_channels = [idx for idx in used_channels if idx < 0 or idx >= in_channels]
    if invalid_channels:
        raise ValueError(
            f"Existem índices de canais inválidos: {invalid_channels[:10]}. "
            f"in_channels={in_channels}."
        )

    if pd.Series(used_channels).duplicated().any():
        raise ValueError("Há canais repetidos em channel_groups.")

    return group_names, group_indices, used_channels


class EarlyFusionSpatialCNN(nn.Module):
    """Todos os canais entram juntos na mesma branch CNN."""
    def __init__(self, in_channels, channel_groups, architecture_config):
        super().__init__()
        group_names, group_indices, used_channels = _validate_channel_groups(in_channels, channel_groups)
        self.group_names = group_names
        self.group_indices = group_indices
        self.used_channels = used_channels
        self.fusion_type = "early"

        cfg = dict(architecture_config)
        self.branch = ArchitectureSpatialBranch(
            in_channels=len(used_channels),
            branch_channels=cfg.get("branch_channels", 32),
            depth=cfg.get("depth", 3),
            block_type=cfg.get("block_type", "conv"),
            pooling_type=cfg.get("pooling_type", "gap2_center"),
            center_radius=cfg.get("center_radius", 1),
            dilation_mode=cfg.get("dilation_mode", "last"),
            se_reduction=cfg.get("se_reduction", 8),
        )
        self.regressor = make_regressor(
            self.branch.out_features,
            regressor_type=cfg.get("regressor_type", "current_mlp"),
            dropout=cfg.get("dropout", 0.3),
        )

    def forward(self, x):
        x = x[:, self.used_channels, :, :]
        features = self.branch(x)
        return self.regressor(features).squeeze(1)


class LateFusionSpatialCNN(nn.Module):
    """Cada grupo/sensor tem branch própria. A fusão acontece no vetor de features final."""
    def __init__(self, in_channels, channel_groups, architecture_config):
        super().__init__()
        group_names, group_indices, _ = _validate_channel_groups(in_channels, channel_groups)
        self.group_names = group_names
        self.group_indices = group_indices
        self.module_keys = [_sanitize_module_key(g) for g in self.group_names]
        self.fusion_type = "late"

        cfg = dict(architecture_config)
        self.branches = nn.ModuleDict()
        total_features = 0

        for group_name, idxs, module_key in zip(self.group_names, self.group_indices, self.module_keys):
            branch = ArchitectureSpatialBranch(
                in_channels=len(idxs),
                branch_channels=cfg.get("branch_channels", 32),
                depth=cfg.get("depth", 3),
                block_type=cfg.get("block_type", "conv"),
                pooling_type=cfg.get("pooling_type", "gap2_center"),
                center_radius=cfg.get("center_radius", 1),
                dilation_mode=cfg.get("dilation_mode", "last"),
                se_reduction=cfg.get("se_reduction", 8),
            )
            self.branches[module_key] = branch
            total_features += branch.out_features

        self.total_features = int(total_features)
        self.regressor = make_regressor(
            self.total_features,
            regressor_type=cfg.get("regressor_type", "current_mlp"),
            dropout=cfg.get("dropout", 0.3),
        )

    def forward(self, x):
        features = []
        for module_key, indices in zip(self.module_keys, self.group_indices):
            x_group = x[:, indices, :, :]
            features.append(self.branches[module_key](x_group))
        features = torch.cat(features, dim=1)
        return self.regressor(features).squeeze(1)


class HybridFusionSpatialCNN(nn.Module):
    """
    Fusão intermédia:
      1) cada sensor/grupo passa por um stem 1x1 próprio;
      2) os mapas de features são concatenados;
      3) blocos CNN comuns extraem padrões espaciais já fundidos;
      4) pooling + regressor.
    """
    def __init__(self, in_channels, channel_groups, architecture_config):
        super().__init__()
        group_names, group_indices, _ = _validate_channel_groups(in_channels, channel_groups)
        self.group_names = group_names
        self.group_indices = group_indices
        self.module_keys = [_sanitize_module_key(g) for g in self.group_names]
        self.fusion_type = "hybrid"

        cfg = dict(architecture_config)
        branch_channels = int(cfg.get("branch_channels", 32))
        stem_channels = int(cfg.get("hybrid_stem_channels", max(branch_channels // 2, 8)))
        depth = int(cfg.get("depth", 3))
        block_type = cfg.get("block_type", "conv")
        dilation_mode = cfg.get("dilation_mode", "last")
        se_reduction = cfg.get("se_reduction", 8)

        self.group_stems = nn.ModuleDict()
        for group_name, idxs, module_key in zip(self.group_names, self.group_indices, self.module_keys):
            self.group_stems[module_key] = ConvBlock(len(idxs), stem_channels, kernel_size=1, dilation=1)

        fused_channels = stem_channels * len(self.group_indices)
        self.fusion_projection = ConvBlock(fused_channels, branch_channels, kernel_size=1, dilation=1)

        dilations = make_dilation_schedule(depth, dilation_mode=dilation_mode)
        self.blocks = nn.Sequential(*[
            make_spatial_block(
                block_type,
                branch_channels,
                branch_channels,
                dilation=d,
                se_reduction=se_reduction,
            )
            for d in dilations
        ])

        self.pooler = SpatialPooler(
            channels=branch_channels,
            pooling_type=cfg.get("pooling_type", "gap2_center"),
            center_radius=cfg.get("center_radius", 1),
        )
        self.regressor = make_regressor(
            self.pooler.out_features,
            regressor_type=cfg.get("regressor_type", "current_mlp"),
            dropout=cfg.get("dropout", 0.3),
        )

    def forward(self, x):
        maps = []
        for module_key, indices in zip(self.module_keys, self.group_indices):
            x_group = x[:, indices, :, :]
            maps.append(self.group_stems[module_key](x_group))
        x = torch.cat(maps, dim=1)
        x = self.fusion_projection(x)
        x = self.blocks(x)
        features = self.pooler(x)
        return self.regressor(features).squeeze(1)


def build_architecture_model(in_channels, channel_groups, architecture_config):
    """Fábrica de modelos. É isto que permite a ablação arquitetural."""
    cfg = dict(architecture_config)
    fusion_type = str(cfg.get("fusion_type", "late")).lower()

    if fusion_type == "early":
        return EarlyFusionSpatialCNN(in_channels, channel_groups, cfg)
    if fusion_type == "late":
        return LateFusionSpatialCNN(in_channels, channel_groups, cfg)
    if fusion_type == "hybrid":
        return HybridFusionSpatialCNN(in_channels, channel_groups, cfg)

    raise ValueError("fusion_type inválido. Usa: early, late ou hybrid.")


def architecture_label(architecture_config):
    """Nome curto e legível para logs, ficheiros e tabelas."""
    cfg = dict(architecture_config)
    parts = [
        f"{cfg.get('fusion_type', 'late')}",
        f"{cfg.get('block_type', 'conv')}",
        f"d{cfg.get('depth', 3)}",
        f"p{cfg.get('pooling_type', 'gap2_center')}",
        f"reg_{cfg.get('regressor_type', 'current_mlp')}",
        f"ch{cfg.get('branch_channels', 32)}",
    ]
    return "_".join(str(p) for p in parts)


def count_parameters(model):
    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Parâmetros totais: {total:,} | treináveis: {trainable:,}")
    return total, trainable
```

```python
# ============================================================
# 6. FUNÇÕES DE MÉTRICAS E TREINO
# ============================================================

def regression_metrics(y_true, y_pred):
    y_true_np = np.array(y_true).flatten()
    y_pred_np = np.array(y_pred).flatten()

    rmse = float(np.sqrt(mean_squared_error(y_true_np, y_pred_np)))
    mae = float(mean_absolute_error(y_true_np, y_pred_np))
    bias = float(np.mean(y_pred_np - y_true_np))

    if len(y_true_np) > 1 and len(np.unique(y_true_np)) > 1:
        r2 = float(r2_score(y_true_np, y_pred_np))
        covariancia = float(np.cov(y_true_np, y_pred_np)[0, 1])
    else:
        r2 = float("nan")
        covariancia = float("nan")

    y_true_mean = float(np.mean(y_true_np))
    pct_rmse = (rmse / y_true_mean) * 100 if y_true_mean > 0 else float("nan")

    return {
        "n": int(len(y_true_np)),
        "RMSE": rmse,
        "%RMSE": pct_rmse,
        "MAE": mae,
        "R2": r2,
        "Bias": bias,
        "Covariancia": covariancia,
    }


def _eval_loader(model, loader):
    """Devolve (y_true, y_pred) em numpy para um loader completo."""
    model.eval()
    all_preds, all_targets = [], []

    with torch.no_grad():
        for bX, by in loader:
            all_preds.append(model(bX.to(device)).cpu().numpy())
            all_targets.append(by.numpy())

    return np.concatenate(all_targets), np.concatenate(all_preds)



def _base_elementwise_loss(y_pred, y_true, loss_config=None):
    """Loss element-wise, sem redução, para permitir pesos por amostra."""
    if loss_config is None:
        loss_config = {"loss_mode": "standard"}

    loss_mode = str(loss_config.get("loss_mode", "standard"))
    base_loss = str(loss_config.get("base_loss", "mse"))

    if loss_mode in ["smooth_l1", "class_weighted_smooth_l1"]:
        base_loss = "smooth_l1"

    if base_loss == "smooth_l1":
        beta = float(loss_config.get("beta", 1.0))
        return F.smooth_l1_loss(y_pred, y_true, reduction="none", beta=beta)

    if base_loss == "mse":
        return (y_pred - y_true) ** 2

    raise ValueError(f"base_loss desconhecida: {base_loss}")


def compute_weighted_mse_loss(y_pred, y_true, loss_config=None):
    """
    Calcula a loss de treino.

    y_true e y_pred estão em metros.

    Modos suportados:
      {"loss_mode": "standard"}
      {"loss_mode": "smooth_l1", "beta": 1.0}
      {"loss_mode": "high_weight", "threshold_m": 20.0, "high_weight": 1.5}
      {"loss_mode": "continuous", "alpha": 1.0, "y_max_train_m": 60.0}
      {"loss_mode": "class_weighted", "class_weights": [...], "class_bins_m": (...)}
      {"loss_mode": "class_weighted_smooth_l1", "class_weights": [...], "class_bins_m": (...), "beta": 1.0}
    """
    if loss_config is None:
        loss_config = {"loss_mode": "standard"}

    loss_mode = str(loss_config.get("loss_mode", "standard"))
    elem_loss = _base_elementwise_loss(y_pred, y_true, loss_config=loss_config)

    weights = torch.ones_like(y_true)

    if loss_mode in ["standard", "smooth_l1"]:
        pass

    elif loss_mode == "high_weight":
        threshold_m = float(loss_config.get("threshold_m", 20.0))
        high_weight = float(loss_config.get("high_weight", 1.5))
        weights = torch.where(
            y_true >= threshold_m,
            torch.full_like(weights, high_weight),
            weights,
        )

    elif loss_mode == "continuous":
        alpha = float(loss_config.get("alpha", 1.0))
        y_max_train_m = float(loss_config.get("y_max_train_m", 1.0))
        y_max_train_m = max(y_max_train_m, 1e-6)
        weights = 1.0 + alpha * torch.clamp(y_true / y_max_train_m, min=0.0, max=1.0)

    elif loss_mode in ["class_weighted", "class_weighted_smooth_l1"]:
        class_weights = torch.as_tensor(
            loss_config.get("class_weights", [1.0]),
            dtype=y_true.dtype,
            device=y_true.device,
        )
        bins_m = loss_config.get("class_bins_m", FIXED_HEIGHT_BINS_M)
        inner_bins = [float(v) for v in bins_m[1:-1]]
        inner_bins_t = torch.as_tensor(inner_bins, dtype=y_true.dtype, device=y_true.device)

        # y_true está em metros. bucketize devolve classes 0..n_classes-1.
        class_idx = torch.bucketize(y_true.detach(), inner_bins_t, right=False)
        class_idx = torch.clamp(class_idx, min=0, max=len(class_weights) - 1)
        weights = class_weights[class_idx]

    else:
        raise ValueError(f"loss_mode desconhecido: {loss_mode}")

    return torch.mean(weights * elem_loss)


def describe_loss_config(loss_config):
    """Texto curto para logs."""
    if loss_config is None:
        return "standard MSE"

    loss_mode = str(loss_config.get("loss_mode", "standard"))

    if loss_mode == "standard":
        return "standard MSE"

    if loss_mode == "smooth_l1":
        return f"SmoothL1Loss | beta={loss_config.get('beta', 1.0)}"

    if loss_mode == "high_weight":
        return (
            f"weighted MSE | y >= {loss_config.get('threshold_m', 20.0)} m "
            f"com peso {loss_config.get('high_weight', 1.5)}"
        )

    if loss_mode == "continuous":
        return (
            f"continuous weighted MSE | alpha={loss_config.get('alpha', 1.0)} | "
            f"y_max_train_m={loss_config.get('y_max_train_m', np.nan):.2f}"
        )

    if loss_mode in ["class_weighted", "class_weighted_smooth_l1"]:
        base = "SmoothL1" if loss_mode.endswith("smooth_l1") else "MSE"
        weights = loss_config.get("class_weights", None)
        weights_txt = "auto" if weights is None else np.round(np.asarray(weights, dtype=float), 3).tolist()
        return f"class-weighted {base} | weights={weights_txt}"

    return str(loss_config)

def train_spacialcnn_cnn(
    model,
    train_loader,
    val_loader,
    model_name,
    epochs=40,
    patience=15,
    lr=3e-4,
    weight_decay=0.0,
    log_every=5,
    loss_config=None,
):
    """
    Treina a LateFusionCNN com early stopping.

    A loss de treino pode ser standard ou ponderada.
    A métrica usada para guardar o melhor checkpoint continua a ser o MSE de validação standard,
    para manter comparação justa entre variantes.
    """
    model = model.to(device)

    if loss_config is None:
        loss_config = {"loss_mode": "standard"}

    print(f"\nLoss de treino: {describe_loss_config(loss_config)}")

    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=5,
    )

    ckpt_path = OUTPUT_DIR / f"ckpt_{model_name}.pth"

    best_val_loss = np.inf
    best_epoch = 0
    epochs_no_impr = 0

    history = {
        "epoch": [],
        "train_loss": [],
        "val_mse": [],
        "val_rmse": [],
        "val_pct_rmse": [],
        "val_mae": [],
        "val_r2": [],
        "lr": [],
    }

    y_val_all = np.concatenate([by.numpy() for _, by in val_loader])
    y_val_mean = float(np.mean(y_val_all))

    header = (
        f"{'Época':>6}  {'Loss_tr':>8}  {'MSE_val':>8}  "
        f"{'RMSE_val':>9}  {'%RMSE':>6}  {'MAE':>7}  {'R²':>6}  "
        f"{'LR':>9}  {'s/ep':>5}  {'ETA':>6}"
    )

    print(f"\n{'─' * len(header)}")
    print(header)
    print(f"{'─' * len(header)}")

    epoch_times = []

    for epoch in range(1, epochs + 1):
        t0 = time.time()

        model.train()
        train_loss_sum = 0.0

        for bX, by in train_loader:
            bX, by = bX.to(device), by.to(device)

            optimizer.zero_grad()
            y_pred = model(bX)
            loss = compute_weighted_mse_loss(y_pred, by, loss_config=loss_config)
            loss.backward()
            optimizer.step()

            train_loss_sum += loss.item() * bX.size(0)

        train_loss = train_loss_sum / len(train_loader.dataset)

        y_true_v, y_pred_v = _eval_loader(model, val_loader)

        val_mse = float(mean_squared_error(y_true_v, y_pred_v))
        val_rmse = float(np.sqrt(val_mse))
        val_pct_rmse = (val_rmse / y_val_mean) * 100 if y_val_mean > 0 else float("nan")
        val_mae = float(mean_absolute_error(y_true_v, y_pred_v))
        val_r2 = float(r2_score(y_true_v, y_pred_v))

        scheduler.step(val_mse)
        current_lr = optimizer.param_groups[0]["lr"]

        elapsed = time.time() - t0
        epoch_times.append(elapsed)

        remaining_epochs = epochs - epoch
        avg_ep_time = np.mean(epoch_times[-10:])
        eta_s = remaining_epochs * avg_ep_time
        eta_str = f"{int(eta_s // 60)}m{int(eta_s % 60):02d}s" if eta_s >= 60 else f"{eta_s:.0f}s"

        improved = val_mse < best_val_loss

        if improved:
            best_val_loss = val_mse
            best_epoch = epoch
            epochs_no_impr = 0
            torch.save(model.state_dict(), ckpt_path)
            flag = " ⭐"
        else:
            epochs_no_impr += 1
            flag = ""

        history["epoch"].append(epoch)
        history["train_loss"].append(train_loss)
        history["val_mse"].append(val_mse)
        history["val_rmse"].append(val_rmse)
        history["val_pct_rmse"].append(val_pct_rmse)
        history["val_mae"].append(val_mae)
        history["val_r2"].append(val_r2)
        history["lr"].append(current_lr)

        if epoch == 1 or epoch % log_every == 0 or improved:
            print(
                f"{epoch:>6}  {train_loss:>8.4f}  {val_mse:>8.4f}  "
                f"{val_rmse:>8.3f}m  {val_pct_rmse:>5.1f}%  "
                f"{val_mae:>6.3f}m  {val_r2:>6.3f}  "
                f"{current_lr:>9.2e}  {elapsed:>4.1f}s  {eta_str:>6}{flag}"
            )

        if epochs_no_impr >= patience:
            print(f"\nEarly stopping na época {epoch} (sem melhoria há {patience} épocas).")
            break

    print(f"{'─' * len(header)}")
    print(
        f"Melhor época: {best_epoch} | "
        f"Melhor val MSE: {best_val_loss:.4f} | "
        f"RMSE: {np.sqrt(best_val_loss):.3f}m"
    )

    model.load_state_dict(torch.load(ckpt_path, map_location=device))
    return model, pd.DataFrame(history)


def evaluate_model(model, val_loader, model_name):
    y_true, y_pred = _eval_loader(model, val_loader)

    metrics = regression_metrics(y_true, y_pred)

    results_df = pd.DataFrame([
        {
            "model": model_name,
            "split": "val",
            **metrics,
        }
    ])

    preds_dict = {
        "val": {
            "y_true": y_true,
            "y_pred": y_pred,
        }
    }

    return results_df, preds_dict
```

```python
# ============================================================
# 7. GRÁFICOS
# ============================================================

PLOTS_DIR = OUTPUT_DIR / "figuras"


def finalize_figure(path=None):
    """Mostra a figura e, se SAVE_PLOTS=True, guarda-a no disco."""
    plt.tight_layout()

    if SAVE_PLOTS and path is not None:
        path = Path(path)
        path.parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(path, dpi=150, bbox_inches="tight")
        print(f"Figura guardada: {path}")

    plt.show()


def clean_label(text):
    """Nome seguro para ficheiros."""
    return str(text).replace(" ", "_").replace("/", "_").replace("|", "_")


def flatten_summary_columns(df):
    """Achata colunas MultiIndex geradas por groupby.agg."""
    df = df.copy()
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = [
            "_".join([str(x) for x in col if str(x) != ""]).strip("_")
            for col in df.columns
        ]
    return df


def display_sorted_summary(summary_df, sort_metric="%RMSE_mean", n=20):
    """Mostra resumo ordenado pelo %RMSE ou RMSE, quando as colunas existem."""
    df = flatten_summary_columns(summary_df)
    if sort_metric not in df.columns:
        candidates = [c for c in df.columns if sort_metric.replace("_mean", "") in c]
        if candidates:
            sort_metric = candidates[0]
    if sort_metric in df.columns:
        df = df.sort_values(sort_metric)
    display(df.head(n))
    return df


def format_height_class_labels(bin_edges, target_scale=100.0):
    """Cria labels legíveis para as classes de altura."""
    labels = {}
    for i in range(len(bin_edges) - 1):
        lo = bin_edges[i] / target_scale
        hi = bin_edges[i + 1] / target_scale
        labels[i] = f"C{i}: {lo:.1f}-{hi:.1f} m"
    return labels


def plot_learning_curves(hist, model_name, split_label):
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    train_col = "train_loss" if "train_loss" in hist.columns else "train_mse"
    axes[0].plot(hist["epoch"], hist[train_col], label="treino")
    axes[0].plot(hist["epoch"], hist["val_mse"], label="validação", linestyle="--")
    axes[0].set(title="Loss treino / MSE validação", xlabel="Época", ylabel="Loss")
    axes[0].legend()
    axes[0].grid(alpha=0.3)

    axes[1].plot(hist["epoch"], hist["val_rmse"], label="RMSE (m)")
    axes[1].plot(hist["epoch"], hist["val_mae"], label="MAE (m)", linestyle="--")
    axes[1].set(title="RMSE & MAE (validação)", xlabel="Época", ylabel="m")
    axes[1].legend()
    axes[1].grid(alpha=0.3)

    axes[2].plot(hist["epoch"], hist["val_pct_rmse"], label="%RMSE")
    ax2r = axes[2].twinx()
    ax2r.plot(hist["epoch"], hist["val_r2"], linestyle=":", label="R²")
    axes[2].set(title="%RMSE & R² (validação)", xlabel="Época", ylabel="%RMSE")
    ax2r.set_ylabel("R²")
    axes[2].legend(loc="upper right")
    ax2r.legend(loc="upper center")
    axes[2].grid(alpha=0.3)

    plt.suptitle(f"{model_name} — {split_label}", fontsize=12, fontweight="bold")
    finalize_figure(PLOTS_DIR / f"curvas_{clean_label(model_name)}_{clean_label(split_label)}.png")


def plot_true_vs_pred(y_true, y_pred, model_name, split_name, split_label):
    mn = min(np.min(y_true), np.min(y_pred))
    mx = max(np.max(y_true), np.max(y_pred))

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)

    plt.figure(figsize=(5.5, 5.5))
    plt.scatter(y_true, y_pred, alpha=0.45, s=12)
    plt.plot([mn, mx], [mn, mx], "--k", linewidth=1)
    plt.xlabel("Altura real (m)")
    plt.ylabel("Altura prevista (m)")
    plt.title(f"{model_name} | {split_name} | {split_label}\nRMSE={rmse:.3f} m  R²={r2:.3f}")
    plt.grid(alpha=0.3)

    finalize_figure(PLOTS_DIR / f"{clean_label(model_name)}_{clean_label(split_name)}_{clean_label(split_label)}_real_vs_previsto.png")


def plot_residuals(y_true, y_pred, model_name, split_name, split_label):
    residuals = y_pred - y_true

    plt.figure(figsize=(6, 4.5))
    plt.scatter(y_pred, residuals, alpha=0.45, s=12)
    plt.axhline(0, linestyle="--", color="k", linewidth=1)
    plt.xlabel("Altura prevista (m)")
    plt.ylabel("Resíduo (previsto - real) (m)")
    plt.title(f"Resíduos | {model_name} | {split_name} | {split_label}")
    plt.grid(alpha=0.3)

    finalize_figure(PLOTS_DIR / f"{clean_label(model_name)}_{clean_label(split_name)}_{clean_label(split_label)}_residuos.png")


def plot_pooled_true_vs_pred(predictions_df, augmentation_mode):
    df = predictions_df[predictions_df["augmentation"] == augmentation_mode].copy()

    plot_true_vs_pred(
        df["y_true_m"].to_numpy(),
        df["y_pred_m"].to_numpy(),
        model_name="LateFusionSpatialCNN",
        split_name="val_pooled",
        split_label=augmentation_mode,
    )


def plot_pooled_residuals(predictions_df, augmentation_mode):
    df = predictions_df[predictions_df["augmentation"] == augmentation_mode].copy()

    plot_residuals(
        df["y_true_m"].to_numpy(),
        df["y_pred_m"].to_numpy(),
        model_name="LateFusionSpatialCNN",
        split_name="val_pooled",
        split_label=augmentation_mode,
    )


def plot_true_pred_histogram(predictions_df, augmentation_mode, n_bins=40):
    """Histograma da distribuição real vs prevista com bins comuns."""
    df = predictions_df[predictions_df["augmentation"] == augmentation_mode].copy()

    y_true = df["y_true_m"].to_numpy()
    y_pred = df["y_pred_m"].to_numpy()

    print(f"Histograma — {augmentation_mode} | N real={len(y_true)} | N previsto={len(y_pred)}")

    min_val = min(np.min(y_true), np.min(y_pred))
    max_val = max(np.max(y_true), np.max(y_pred))
    bins = np.linspace(min_val, max_val, n_bins + 1)

    plt.figure(figsize=(7, 4.5))
    plt.hist(y_true, bins=bins, alpha=0.55, label="Real")
    plt.hist(y_pred, bins=bins, alpha=0.55, label="Previsto")
    plt.xlabel("Altura (m)")
    plt.ylabel("Nº de amostras")
    plt.title(f"Distribuição de alturas reais e previstas — {augmentation_mode}")
    plt.legend()
    plt.grid(alpha=0.3)

    finalize_figure(PLOTS_DIR / f"hist_real_vs_previsto_{augmentation_mode}.png")

def plot_metric_by_region(metrics_by_region, augmentation_mode, metric="RMSE"):
    """Barplot de uma métrica por região."""
    df = metrics_by_region[metrics_by_region["augmentation"] == augmentation_mode].copy()
    df = df.sort_values(metric, ascending=False)

    plt.figure(figsize=(8, 4.8))
    plt.bar(df["region"], df[metric])
    plt.xlabel("Região")
    plt.ylabel(metric)
    plt.title(f"{metric} por região — {augmentation_mode}")
    plt.xticks(rotation=30, ha="right")
    plt.grid(axis="y", alpha=0.3)

    finalize_figure(PLOTS_DIR / f"{metric.lower()}_por_regiao_{augmentation_mode}.png")


def plot_metric_by_height_class(metrics_by_height_class, augmentation_mode, metric="RMSE", height_class_labels=None):
    """Barplot de uma métrica por classe de altura."""
    df = metrics_by_height_class[metrics_by_height_class["augmentation"] == augmentation_mode].copy()
    df = df.sort_values("height_class")

    xlabels = df["height_class"].astype(str).tolist()
    if height_class_labels is not None:
        xlabels = [height_class_labels.get(int(c), str(c)) for c in df["height_class"]]

    plt.figure(figsize=(8, 4.8))
    plt.bar(xlabels, df[metric])
    plt.xlabel("Classe de altura")
    plt.ylabel(metric)
    plt.title(f"{metric} por classe de altura — {augmentation_mode}")
    plt.xticks(rotation=30, ha="right")
    plt.grid(axis="y", alpha=0.3)

    finalize_figure(PLOTS_DIR / f"{metric.lower()}_por_classe_altura_{augmentation_mode}.png")


def plot_residual_boxplot_by_height_class(predictions_df, augmentation_mode, height_class_labels=None):
    """Boxplot dos resíduos por classe de altura."""
    df = predictions_df[predictions_df["augmentation"] == augmentation_mode].copy()
    classes = sorted(df["height_class"].unique())
    data = [df[df["height_class"] == c]["residual_m"].to_numpy() for c in classes]

    labels = [str(c) for c in classes]
    if height_class_labels is not None:
        labels = [height_class_labels.get(int(c), str(c)) for c in classes]

    plt.figure(figsize=(8, 4.8))
    plt.boxplot(data, labels=labels, showfliers=False)
    plt.axhline(0, linestyle="--", color="k", linewidth=1)
    plt.xlabel("Classe de altura")
    plt.ylabel("Resíduo (previsto - real) (m)")
    plt.title(f"Distribuição dos resíduos por classe de altura — {augmentation_mode}")
    plt.xticks(rotation=30, ha="right")
    plt.grid(axis="y", alpha=0.3)

    finalize_figure(PLOTS_DIR / f"boxplot_residuos_por_classe_altura_{augmentation_mode}.png")


def plot_residual_boxplot_by_region(predictions_df, augmentation_mode):
    """Boxplot dos resíduos por região."""
    df = predictions_df[predictions_df["augmentation"] == augmentation_mode].copy()
    regions = sorted(df["region"].unique())
    data = [df[df["region"] == r]["residual_m"].to_numpy() for r in regions]

    plt.figure(figsize=(8, 4.8))
    plt.boxplot(data, labels=regions, showfliers=False)
    plt.axhline(0, linestyle="--", color="k", linewidth=1)
    plt.xlabel("Região")
    plt.ylabel("Resíduo (previsto - real) (m)")
    plt.title(f"Distribuição dos resíduos por região — {augmentation_mode}")
    plt.xticks(rotation=30, ha="right")
    plt.grid(axis="y", alpha=0.3)

    finalize_figure(PLOTS_DIR / f"boxplot_residuos_por_regiao_{augmentation_mode}.png")


def plot_cv_metric_evolution(all_histories, augmentation_mode, metric="val_rmse"):
    """
    Mostra a evolução média de uma métrica de validação ao longo das épocas.
    Como alguns folds podem parar mais cedo, calcula a média apenas das épocas disponíveis.
    """
    rows = []
    for (mode, fold), hist in all_histories.items():
        if mode != augmentation_mode:
            continue
        df_hist = hist[["epoch", metric]].copy()
        df_hist["fold"] = fold
        rows.append(df_hist)

    if not rows:
        return

    hist_df = pd.concat(rows, ignore_index=True)
    summary = (
        hist_df
        .groupby("epoch")[metric]
        .agg(["mean", "std", "count"])
        .reset_index()
    )

    plt.figure(figsize=(7, 4.5))
    plt.plot(summary["epoch"], summary["mean"], label=f"média {metric}")

    lower = summary["mean"] - summary["std"].fillna(0)
    upper = summary["mean"] + summary["std"].fillna(0)
    plt.fill_between(summary["epoch"], lower, upper, alpha=0.2, label="± desvio padrão")

    plt.xlabel("Época")
    plt.ylabel(metric)
    plt.title(f"Evolução média de {metric} nos folds — {augmentation_mode}")
    plt.legend()
    plt.grid(alpha=0.3)

    finalize_figure(PLOTS_DIR / f"evolucao_media_{metric}_{augmentation_mode}.png")


def plot_all_final_figures(
    predictions_df,
    metrics_by_region,
    metrics_by_height_class,
    all_histories,
    bin_edges,
):
    """Gera os gráficos finais mais úteis para análise e tese."""
    height_class_labels = format_height_class_labels(bin_edges, target_scale=TARGET_SCALE)

    for augmentation_mode in sorted(predictions_df["augmentation"].unique()):
        print(f"\n=== Gráficos finais — {augmentation_mode} ===")

        plot_pooled_true_vs_pred(predictions_df, augmentation_mode)
        plot_pooled_residuals(predictions_df, augmentation_mode)
        plot_true_pred_histogram(predictions_df, augmentation_mode)

        plot_metric_by_region(metrics_by_region, augmentation_mode, metric="RMSE")
        plot_metric_by_region(metrics_by_region, augmentation_mode, metric="R2")
        plot_metric_by_height_class(metrics_by_height_class, augmentation_mode, metric="RMSE", height_class_labels=height_class_labels)
        plot_metric_by_height_class(metrics_by_height_class, augmentation_mode, metric="Bias", height_class_labels=height_class_labels)

        plot_residual_boxplot_by_region(predictions_df, augmentation_mode)
        plot_residual_boxplot_by_height_class(predictions_df, augmentation_mode, height_class_labels=height_class_labels)

        plot_cv_metric_evolution(all_histories, augmentation_mode, metric="val_rmse")
        plot_cv_metric_evolution(all_histories, augmentation_mode, metric="val_r2")
```

```python
# ============================================================
# 8. PIPELINE — LATEFUSION CENTER-AWARE 11x11 + STANDARD ENSEMBLE + HIGH-HEIGHT WEIGHTED LOSS
# ============================================================

# ------------------------------------------------------------
# Classes de altura e estratificação
# ------------------------------------------------------------

def make_height_classes(y, n_bins=5):
    """
    Cria classes de altura por quantis.

    Com n_bins=5:
      classe 0 = alturas mais baixas
      classe 4 = alturas mais altas
    """
    y = np.asarray(y, dtype=float)

    if len(y) == 0:
        raise ValueError("y está vazio.")

    if len(np.unique(y)) < 2 or n_bins <= 1:
        classes = np.zeros(len(y), dtype=int)
        bins = np.array([np.min(y), np.max(y)])
        return classes, bins

    classes, bins = pd.qcut(
        pd.Series(y),
        q=n_bins,
        labels=False,
        duplicates="drop",
        retbins=True,
    )

    classes = classes.astype(int).to_numpy()
    return classes, bins


def choose_region_height_strata_for_cv(y, regions, n_splits=5, max_bins=5):
    """
    Cria estratos combinados região + classe_altura.

    Isto tenta garantir que cada fold tem:
      - proporção semelhante de pontos de cada região;
      - proporção semelhante de classes de altura;
      - proporção semelhante de cada combinação região/classe_altura.
    """
    y = np.asarray(y, dtype=float)
    regions = np.asarray(regions)

    if len(y) < n_splits:
        raise ValueError(
            f"Dataset tem apenas {len(y)} amostras. "
            f"Não dá para fazer {n_splits} folds."
        )

    max_bins_possible = min(max_bins, len(np.unique(y)), len(y))

    for n_bins in range(max_bins_possible, 0, -1):
        height_classes, bin_edges = make_height_classes(y, n_bins=n_bins)

        strata = np.array([
            f"{region}__h{height_class}"
            for region, height_class in zip(regions, height_classes)
        ])

        counts = pd.Series(strata).value_counts()

        if counts.min() >= n_splits:
            if n_bins < max_bins:
                print(
                    f"AVISO global: reduzi as classes de altura de "
                    f"{max_bins} para {n_bins}, porque havia combinações "
                    f"região/classe_altura com menos de {n_splits} amostras."
                )

            return height_classes, strata, bin_edges

    min_region_count = pd.Series(regions).value_counts().min()

    raise ValueError(
        "Não foi possível criar estratos região + altura para "
        f"{n_splits} folds. A região mais pequena tem {min_region_count} amostras."
    )


def print_height_bins(bin_edges, target_scale=100.0):
    """Mostra os limites das classes de altura em metros."""
    print("\nLimites das classes de altura:")

    for i in range(len(bin_edges) - 1):
        lo = bin_edges[i] / target_scale
        hi = bin_edges[i + 1] / target_scale
        print(f"  Classe {i}: {lo:.2f} m até {hi:.2f} m")


def make_fixed_height_classes(y, bins_m=None, target_scale=100.0):
    """
    Cria classes fixas de altura em metros.

    Por defeito:
      classe 0 = 0-5 m
      classe 1 = 5-10 m
      classe 2 = 10-15 m
      classe 3 = 15-20 m
      classe 4 = >20 m

    A entrada y está em centímetros, como no dataset original.
    """
    if bins_m is None:
        bins_m = FIXED_HEIGHT_BINS_M

    y_m = np.asarray(y, dtype=float) / target_scale
    bins_m = np.asarray(bins_m, dtype=float)

    if not np.isinf(bins_m[-1]):
        raise ValueError("O último limite de bins_m deve ser np.inf.")

    classes = np.digitize(y_m, bins_m[1:-1], right=False).astype(int)

    labels = []
    for i in range(len(bins_m) - 1):
        lo, hi = bins_m[i], bins_m[i + 1]
        if np.isinf(hi):
            labels.append(f">{lo:g} m")
        else:
            labels.append(f"{lo:g}-{hi:g} m")

    return classes, bins_m, labels


def print_fixed_height_bins(bins_m, labels):
    print("\nClasses fixas de altura usadas nas métricas e no sampler:")
    for i, label in enumerate(labels):
        print(f"  Classe {i}: {label}")


def choose_region_fixed_height_strata_for_cv(y, regions, n_splits=5):
    """
    Cria estratos região + classe de altura fixa.

    Se alguma combinação região/classe fixa tiver menos de n_splits amostras,
    usa a versão por quantis apenas para a divisão das folds.
    As métricas e o sampler continuam a usar as classes fixas.
    """
    fixed_classes, bins_m, labels = make_fixed_height_classes(
        y,
        bins_m=FIXED_HEIGHT_BINS_M,
        target_scale=TARGET_SCALE,
    )

    fixed_strata = np.array([
        f"{region}__h{height_class}"
        for region, height_class in zip(regions, fixed_classes)
    ])

    counts = pd.Series(fixed_strata).value_counts()

    if counts.min() >= n_splits:
        print("\nCV: a usar estratificação por região + classes fixas de altura.")
        return fixed_classes, fixed_strata, bins_m, labels, "fixed"

    print("\nAVISO: algumas combinações região + classe fixa têm menos de "
          f"{n_splits} amostras.")
    print("A usar quantis apenas para criar as folds. Métricas e sampler continuam com classes fixas.")

    quantile_classes, quantile_strata, _ = choose_region_height_strata_for_cv(
        y,
        regions,
        n_splits=n_splits,
        max_bins=N_HEIGHT_BINS,
    )

    return fixed_classes, quantile_strata, bins_m, labels, "quantile_for_cv"


# ------------------------------------------------------------
# WeightedRandomSampler
# ------------------------------------------------------------

def make_inverse_frequency_sample_weights(classes):
    """
    Cria pesos inversamente proporcionais à frequência de cada classe.

    Exemplo:
      classe rara  -> peso maior
      classe comum -> peso menor

    Isto não altera a loss. Apenas altera a probabilidade de cada amostra aparecer nos batches.
    """
    classes = np.asarray(classes)
    counts = pd.Series(classes).value_counts().to_dict()

    weights = np.array([
        1.0 / counts[c]
        for c in classes
    ], dtype=np.float64)

    # Normalizar para média 1 melhora a leitura, mas não altera o sampling relativo.
    weights = weights / weights.mean()
    return weights


def print_sampler_diagnostics(train_height_classes, weights):
    """Mostra a distribuição original e os pesos médios por classe no treino."""
    df = pd.DataFrame({
        "height_class": train_height_classes,
        "weight": weights,
    })

    diag = (
        df.groupby("height_class")
        .agg(
            n=("height_class", "size"),
            mean_weight=("weight", "mean"),
            min_weight=("weight", "min"),
            max_weight=("weight", "max"),
        )
        .reset_index()
    )

    print("\nDiagnóstico do WeightedRandomSampler:")
    display(diag)
    return diag


def build_loaders_from_indices(
    X_all,
    y_all,
    train_idx,
    val_idx,
    train_transform=None,
    train_height_classes=None,
    sampling_mode="standard",
    seed=42,
):
    """
    Normaliza sem leakage e cria DataLoaders.

    A média e o desvio padrão são calculados apenas com o treino do fold.
    A validação nunca recebe data augmentation.

    sampling_mode:
      - "standard": shuffle normal.
      - "weighted_sampler": WeightedRandomSampler por classe de altura.
    """
    mean, std = get_cached_channel_stats(X_all, train_idx)

    y_train = y_all[train_idx] / TARGET_SCALE
    y_val = y_all[val_idx] / TARGET_SCALE

    print(
        f"\ny_train: média={y_train.mean():.3f}m | std={y_train.std():.3f}m | "
        f"min={y_train.min():.2f}m | max={y_train.max():.2f}m"
    )
    print(
        f"y_val:   média={y_val.mean():.3f}m | std={y_val.std():.3f}m | "
        f"min={y_val.min():.2f}m | max={y_val.max():.2f}m"
    )

    train_ds = NormalizedIndexedFCHDataset(
        X_all=X_all,
        y_all=y_all,
        indices=train_idx,
        mean=mean,
        std=std,
        transform=train_transform,
        target_scale=TARGET_SCALE,
    )
    val_ds = NormalizedIndexedFCHDataset(
        X_all=X_all,
        y_all=y_all,
        indices=val_idx,
        mean=mean,
        std=std,
        transform=None,
        target_scale=TARGET_SCALE,
    )

    generator = torch.Generator()
    generator.manual_seed(seed)

    sampler = None
    shuffle = True

    if sampling_mode == "weighted_sampler":
        if train_height_classes is None:
            train_height_classes, _ = make_height_classes(y_train, n_bins=N_HEIGHT_BINS)

        sample_weights = make_inverse_frequency_sample_weights(train_height_classes)
        print_sampler_diagnostics(train_height_classes, sample_weights)

        sampler = WeightedRandomSampler(
            weights=torch.as_tensor(sample_weights, dtype=torch.double),
            num_samples=len(sample_weights),
            replacement=True,
            generator=generator,
        )
        shuffle = False

    elif sampling_mode == "standard":
        sampler = None
        shuffle = True

    else:
        raise ValueError("sampling_mode deve ser 'standard' ou 'weighted_sampler'.")

    train_loader = DataLoader(
        train_ds,
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        sampler=sampler,
        drop_last=(len(train_ds) % BATCH_SIZE == 1),
        generator=generator if shuffle else None,
        num_workers=DATALOADER_NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
    )

    val_loader = DataLoader(
        val_ds,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=DATALOADER_NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
    )

    return train_loader, val_loader


# ------------------------------------------------------------
# Loss ponderada por classe de altura
# ------------------------------------------------------------

def make_class_weights_from_train_y(y_train_cm, bins_m=None, target_scale=100.0, power=1.0):
    """
    Cria pesos inversos à frequência por classe fixa de altura.

    Os pesos são normalizados para que o peso médio das amostras de treino seja 1.
    Assim a escala da loss não muda drasticamente.
    """
    if bins_m is None:
        bins_m = FIXED_HEIGHT_BINS_M

    train_classes, _, _ = make_fixed_height_classes(
        y_train_cm,
        bins_m=bins_m,
        target_scale=target_scale,
    )

    n_classes = len(bins_m) - 1
    counts = pd.Series(train_classes).value_counts().to_dict()

    raw_weights = np.ones(n_classes, dtype=np.float32)
    for c in range(n_classes):
        count_c = counts.get(c, 0)
        if count_c > 0:
            raw_weights[c] = (len(train_classes) / (n_classes * count_c)) ** float(power)
        else:
            raw_weights[c] = 1.0

    sample_weights = np.array([raw_weights[c] for c in train_classes], dtype=np.float32)
    mean_weight = float(sample_weights.mean()) if len(sample_weights) else 1.0
    if mean_weight > 0:
        raw_weights = raw_weights / mean_weight

    return raw_weights.tolist()


def prepare_loss_config_for_fold(loss_config, y_train_cm):
    """Preenche parâmetros dependentes do fold, sem usar validação."""
    cfg = dict(loss_config or {"loss_mode": "standard"})
    loss_mode = str(cfg.get("loss_mode", "standard"))

    if loss_mode in ["class_weighted", "class_weighted_smooth_l1"]:
        bins_m = tuple(cfg.get("class_bins_m", tuple(FIXED_HEIGHT_BINS_M)))
        power = float(cfg.get("class_weight_power", 1.0))
        cfg["class_bins_m"] = bins_m
        cfg["class_weights"] = make_class_weights_from_train_y(
            y_train_cm,
            bins_m=bins_m,
            target_scale=TARGET_SCALE,
            power=power,
        )

    if loss_mode == "continuous" and "y_max_train_m" not in cfg:
        cfg["y_max_train_m"] = float(np.nanmax(np.asarray(y_train_cm, dtype=float)) / TARGET_SCALE)

    return cfg

# ------------------------------------------------------------
# Métricas agrupadas
# ------------------------------------------------------------

def grouped_regression_metrics(predictions_df, group_cols):
    """Calcula métricas agrupadas usando previsões pooled."""
    rows = []

    for group_values, df_group in predictions_df.groupby(group_cols):
        if not isinstance(group_values, tuple):
            group_values = (group_values,)

        metrics = regression_metrics(
            df_group["y_true_m"].to_numpy(),
            df_group["y_pred_m"].to_numpy(),
        )

        row = {col: value for col, value in zip(group_cols, group_values)}
        row.update(metrics)
        rows.append(row)

    return pd.DataFrame(rows)


def summarize_fold_metrics(fold_metrics_df):
    metric_cols = ["RMSE", "%RMSE", "MAE", "R2", "Bias", "Covariancia"]

    summary = (
        fold_metrics_df
        .groupby(["model_variant", "sampling_mode", "loss_mode", "augmentation", "seed"])[metric_cols]
        .agg(["mean", "std"])
        .reset_index()
    )

    return summary


# ------------------------------------------------------------
# Helpers para ablations
# ------------------------------------------------------------

def filter_channel_groups(channel_groups, groups_to_keep=None):
    """
    Filtra os grupos de canais usados pelo modelo.

    groups_to_keep:
      - None ou "all": usa todos os grupos detetados;
      - lista de nomes, por exemplo ["S1_SAR"] ou ["S2_OPTICAL"].

    O tensor X continua a ter todos os canais, mas o modelo só lê os índices
    presentes nos grupos selecionados.
    """
    if groups_to_keep is None or groups_to_keep == "all":
        return OrderedDict((group, list(idxs)) for group, idxs in channel_groups.items())

    if isinstance(groups_to_keep, str):
        groups_to_keep = [groups_to_keep]

    filtered = OrderedDict()
    missing = []
    for group in groups_to_keep:
        if group in channel_groups:
            filtered[group] = list(channel_groups[group])
        else:
            missing.append(group)

    if missing:
        print(f"AVISO: grupos pedidos mas não encontrados: {missing}")

    if len(filtered) == 0:
        raise ValueError(
            f"Nenhum dos grupos pedidos existe em CHANNEL_GROUPS: {groups_to_keep}. "
            f"Grupos disponíveis: {list(channel_groups.keys())}"
        )

    return filtered


def describe_channel_groups(channel_groups):
    """Texto curto com grupos ativos e número total de canais."""
    parts = [f"{group}:{len(idxs)}" for group, idxs in channel_groups.items()]
    return "+".join(parts)


def count_active_channels(channel_groups):
    return int(sum(len(idxs) for idxs in channel_groups.values()))


# ------------------------------------------------------------
# Treino de um fold
# ------------------------------------------------------------

def run_fold_for_variant(
    X_all,
    y_all,
    regions_all,
    pids_all,
    height_classes_all,
    sampler_classes_all,
    train_idx,
    val_idx,
    fold,
    model_variant,
    sampling_mode,
    seed,
    train_transform=None,
    loss_config=None,
    channel_groups=None,
    architecture_config=None,
    ablation_type="architecture",
):
    """Treina e avalia um fold para uma variante arquitetural."""
    if loss_config is None:
        loss_config = {"loss_mode": "standard"}
    if channel_groups is None:
        channel_groups = CHANNEL_GROUPS
    if architecture_config is None:
        architecture_config = DEFAULT_ARCH_CONFIG

    architecture_config = dict(architecture_config)
    # Alguns modos de loss precisam de pesos calculados só no treino do fold.
    loss_config = prepare_loss_config_for_fold(loss_config, y_all[train_idx])
    active_groups_label = describe_channel_groups(channel_groups)
    n_active_channels = count_active_channels(channel_groups)

    fusion_type = architecture_config.get("fusion_type", "late")
    block_type = architecture_config.get("block_type", "conv")
    depth = int(architecture_config.get("depth", 3))
    pooling_type = architecture_config.get("pooling_type", "gap2_center")
    regressor_type = architecture_config.get("regressor_type", "current_mlp")
    branch_channels = int(architecture_config.get("branch_channels", 32))
    center_radius = int(architecture_config.get("center_radius", 1))
    dilation_mode = architecture_config.get("dilation_mode", "last")

    print("\n" + "=" * 80)
    print(
        f" Patch {PATCH_SIZE_TO_USE}x{PATCH_SIZE_TO_USE} | "
        f"{model_variant} | seed={seed} | fold={fold}/{N_SPLITS}"
    )
    print(f"Loss config: {describe_loss_config(loss_config)}")
    print(
        f"Arquitetura: fusion={fusion_type} | block={block_type} | depth={depth} | "
        f"pooling={pooling_type} | regressor={regressor_type} | channels={branch_channels}"
    )
    print(f"Center radius: {center_radius} -> janela central {(2 * center_radius + 1)}x{(2 * center_radius + 1)}")
    print(f"Dilation mode: {dilation_mode}")
    print(f"Grupos ativos: {active_groups_label} | canais ativos: {n_active_channels}/{X_all.shape[1]}")
    print("=" * 80)
    print(f"Treino: {len(train_idx)} amostras")
    print(f"Validação: {len(val_idx)} amostras")
    print(f"Regiões treino: {sorted(set(regions_all[train_idx]))}")
    print(f"Regiões validação: {sorted(set(regions_all[val_idx]))}")

    set_seed(seed)

    train_loader, val_loader = build_loaders_from_indices(
        X_all,
        y_all,
        train_idx,
        val_idx,
        train_transform=train_transform,
        train_height_classes=sampler_classes_all[train_idx] if sampler_classes_all is not None else height_classes_all[train_idx],
        sampling_mode=sampling_mode,
        seed=seed,
    )

    num_channels = X_all.shape[1]
    model = build_architecture_model(
        in_channels=num_channels,
        channel_groups=channel_groups,
        architecture_config=architecture_config,
    )

    architecture_name = model.__class__.__name__
    print(f"\nArquitetura PyTorch: {architecture_name}")
    n_params, n_trainable_params = count_parameters(model)

    safe_variant = re.sub(r"[^A-Za-z0-9_]+", "_", str(model_variant))
    model_name = (
        f"{architecture_name}_patch{PATCH_SIZE_TO_USE}x{PATCH_SIZE_TO_USE}_"
        f"{safe_variant}_seed{seed}_fold{fold}"
    )

    model, history = train_spacialcnn_cnn(
        model,
        train_loader,
        val_loader,
        model_name=model_name,
        epochs=EPOCHS,
        patience=PATIENCE,
        lr=LR,
        weight_decay=WEIGHT_DECAY,
        log_every=LOG_EVERY,
        loss_config=loss_config,
    )

    results_df, preds = evaluate_model(
        model,
        val_loader,
        model_name=model_variant,
    )

    common_meta = {
        "architecture": architecture_name,
        "ablation_type": ablation_type,
        "fusion_type": fusion_type,
        "block_type": block_type,
        "depth": depth,
        "pooling_type": pooling_type,
        "regressor_type": regressor_type,
        "branch_channels": branch_channels,
        "center_radius": center_radius,
        "center_window": f"{2 * center_radius + 1}x{2 * center_radius + 1}",
        "dilation_mode": dilation_mode,
        "active_groups": active_groups_label,
        "n_active_channels": n_active_channels,
        "n_params": n_params,
        "n_trainable_params": n_trainable_params,
        "patch_size": PATCH_SIZE_TO_USE,
        "patch_area_m": PATCH_SIZE_TO_USE * 25,
        "augmentation": AUGMENTATION_MODE,
        "sampling_mode": sampling_mode,
        "sampler_label": globals().get("CURRENT_SAMPLER_LABEL", "none"),
        "loss_mode": loss_config.get("loss_mode", "standard"),
        "high_threshold_m": loss_config.get("threshold_m", np.nan),
        "high_weight": loss_config.get("high_weight", np.nan),
        "continuous_alpha": loss_config.get("alpha", np.nan),
        "seed": seed,
        "fold": fold,
        "n_train": len(train_idx),
        "n_val": len(val_idx),
    }

    for col, val in common_meta.items():
        results_df[col] = val

    val_row = results_df[results_df["split"] == "val"].iloc[0]
    print(
        f"\n>>> Validação Final | RMSE={val_row['RMSE']:.3f}m | "
        f"%RMSE={val_row['%RMSE']:.1f}% | MAE={val_row['MAE']:.3f}m | "
        f"R²={val_row['R2']:.3f} | Bias={val_row['Bias']:.3f}m"
    )

    y_true_val = preds["val"]["y_true"]
    y_pred_val = preds["val"]["y_pred"]

    predictions_df = pd.DataFrame({
        **{k: v for k, v in common_meta.items() if k not in ["n_train", "n_val"]},
        "model_variant": model_variant,
        "region": regions_all[val_idx],
        "point_id": pids_all[val_idx],
        "height_class": height_classes_all[val_idx],
        "y_true_m": y_true_val,
        "y_pred_m": y_pred_val,
    })

    predictions_df["height_class_label"] = predictions_df["height_class"].map(HEIGHT_CLASS_LABEL_MAP)
    predictions_df["residual_m"] = predictions_df["y_pred_m"] - predictions_df["y_true_m"]
    predictions_df["abs_error_m"] = predictions_df["residual_m"].abs()

    for col, val in common_meta.items():
        history[col] = val
    history["model_variant"] = model_variant

    del model, train_loader, val_loader
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return results_df, history, predictions_df


# ------------------------------------------------------------
# Ensemble
# ------------------------------------------------------------

def build_ensemble_predictions(individual_predictions_df, model_variant, sampling_mode, seeds):
    """Cria previsões ensemble por média das previsões das várias seeds."""
    df = individual_predictions_df[
        (individual_predictions_df["model_variant"] == model_variant)
        & (individual_predictions_df["sampling_mode"] == sampling_mode)
        & (individual_predictions_df["seed"].isin(seeds))
    ].copy()

    if df.empty:
        return pd.DataFrame()

    group_cols = [
        "augmentation",
        "fold",
        "patch_size",
        "patch_area_m",
        "region",
        "point_id",
        "height_class",
        "height_class_label",
        "y_true_m",
    ]

    ensemble_df = df.groupby(group_cols, as_index=False)["y_pred_m"].mean()
    ensemble_df["model_variant"] = f"{model_variant}_ensemble{len(seeds)}"
    ensemble_df["sampling_mode"] = sampling_mode

    meta_cols = [
        "architecture", "ablation_type", "fusion_type", "block_type", "depth",
        "pooling_type", "regressor_type", "branch_channels", "center_radius",
        "center_window", "dilation_mode", "active_groups", "n_active_channels",
        "n_params", "n_trainable_params", "loss_mode", "high_threshold_m",
        "high_weight", "continuous_alpha",
    ]
    for col in meta_cols:
        if col in df.columns:
            vals = df[col].dropna().unique()
            ensemble_df[col] = vals[0] if len(vals) > 0 else np.nan

    ensemble_df["seed"] = "ensemble"
    ensemble_df["residual_m"] = ensemble_df["y_pred_m"] - ensemble_df["y_true_m"]
    ensemble_df["abs_error_m"] = ensemble_df["residual_m"].abs()

    ordered_cols = [
        "architecture", "ablation_type", "fusion_type", "block_type", "depth",
        "pooling_type", "regressor_type", "branch_channels", "center_radius",
        "center_window", "dilation_mode", "active_groups", "n_active_channels",
        "n_params", "n_trainable_params", "model_variant", "sampling_mode",
        "loss_mode", "high_threshold_m", "high_weight", "continuous_alpha",
        "augmentation", "seed", "fold", "patch_size", "patch_area_m", "region",
        "point_id", "height_class", "height_class_label", "y_true_m",
        "y_pred_m", "residual_m", "abs_error_m",
    ]
    ordered_cols = [c for c in ordered_cols if c in ensemble_df.columns]
    return ensemble_df[ordered_cols]


def metrics_by_fold_from_predictions(predictions_df):
    group_cols = [
        "ablation_type", "architecture", "fusion_type", "block_type", "depth",
        "pooling_type", "regressor_type", "branch_channels", "center_radius",
        "center_window", "dilation_mode", "active_groups", "n_active_channels",
        "n_params", "n_trainable_params", "model_variant", "sampling_mode",
        "loss_mode", "augmentation", "seed", "fold",
    ]
    group_cols = [c for c in group_cols if c in predictions_df.columns]
    return grouped_regression_metrics(predictions_df, group_cols=group_cols)


# ------------------------------------------------------------
# Gráficos finais focados
# ------------------------------------------------------------

def plot_metric_comparison(fold_metrics_df, metric="%RMSE"):
    """Compara modelos por RMSE, %RMSE, MAE, R2 ou Bias."""
    plot_df = (
        fold_metrics_df
        .groupby(["model_variant", "seed"], dropna=False)[metric]
        .agg(["mean", "std"])
        .reset_index()
    )

    ascending = False if metric == "R2" else True
    plot_df = plot_df.sort_values("mean", ascending=ascending)
    plot_df["label"] = plot_df["model_variant"].astype(str) + "\nseed=" + plot_df["seed"].astype(str)

    plt.figure(figsize=(max(10, 0.55 * len(plot_df)), 5.5))
    x = np.arange(len(plot_df))
    plt.bar(x, plot_df["mean"], yerr=plot_df["std"], capsize=4)
    plt.xticks(x, plot_df["label"], rotation=55, ha="right")
    plt.ylabel(f"{metric} médio nas folds")
    plt.title(f"Testes focados — comparação de {metric}")
    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()

    safe_metric = re.sub(r"[^A-Za-z0-9_]+", "_", metric)
    fig_path = PLOTS_DIR / f"comparacao_{safe_metric}_testes_focados.png"
    plt.savefig(fig_path, dpi=150, bbox_inches="tight")
    print(f"Figura guardada: {fig_path}")
    plt.show()


def plot_error_distribution_by_height(predictions_df, top_models_df, metric_name="abs_error_m"):
    """Boxplot do erro por classe de altura para os melhores modelos."""
    model_seed_pairs = set(zip(top_models_df["model_variant"].astype(str), top_models_df["seed"].astype(str)))
    df = predictions_df[
        predictions_df.apply(lambda r: (str(r["model_variant"]), str(r["seed"])) in model_seed_pairs, axis=1)
    ].copy()

    if df.empty:
        print("Sem dados para boxplot por classe de altura.")
        return

    class_order_df = (
        df[["height_class", "height_class_label"]]
        .drop_duplicates()
        .sort_values("height_class")
    )
    labels = class_order_df["height_class_label"].tolist()
    models = top_models_df["model_variant"].astype(str).tolist()

    for model_variant in models:
        df_m = df[df["model_variant"].astype(str) == str(model_variant)]
        if df_m.empty:
            continue

        data = [df_m[df_m["height_class_label"] == lab][metric_name].dropna().to_numpy() for lab in labels]
        plt.figure(figsize=(8, 4.8))
        plt.boxplot(data, labels=labels, showfliers=False)
        plt.xlabel("Classe de altura")
        plt.ylabel("Erro absoluto (m)" if metric_name == "abs_error_m" else metric_name)
        plt.title(f"Distribuição do erro por classe — {model_variant}")
        plt.xticks(rotation=30, ha="right")
        plt.grid(axis="y", alpha=0.3)
        plt.tight_layout()

        safe_model = re.sub(r"[^A-Za-z0-9_]+", "_", str(model_variant))
        fig_path = PLOTS_DIR / f"boxplot_erro_por_classe_{safe_model}.png"
        plt.savefig(fig_path, dpi=150, bbox_inches="tight")
        print(f"Figura guardada: {fig_path}")
        plt.show()

# ------------------------------------------------------------
# Gráficos simples finais
# ------------------------------------------------------------

def plot_rmse_comparison(fold_metrics_df):
    plot_df = (
        fold_metrics_df
        .groupby(["model_variant", "seed"], dropna=False)["RMSE"]
        .agg(["mean", "std"])
        .reset_index()
    )

    plot_df["label"] = plot_df["model_variant"].astype(str) + "\nseed=" + plot_df["seed"].astype(str)
    plot_df = plot_df.sort_values("mean")

    plt.figure(figsize=(max(10, 0.55 * len(plot_df)), 5.5))
    x = np.arange(len(plot_df))
    plt.bar(x, plot_df["mean"], yerr=plot_df["std"], capsize=4)
    plt.xticks(x, plot_df["label"], rotation=55, ha="right")
    plt.ylabel("RMSE médio nas folds (m)")
    plt.title("Testes focados — comparação de RMSE")
    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()

    fig_path = PLOTS_DIR / "comparacao_rmse_arquiteturas.png"
    plt.savefig(fig_path, dpi=150, bbox_inches="tight")
    print(f"Figura guardada: {fig_path}")
    plt.show()


def plot_bias_by_height_class(predictions_df):
    metric_df = grouped_regression_metrics(
        predictions_df,
        group_cols=["model_variant", "seed", "height_class"],
    )

    plt.figure(figsize=(9, 5.5))

    for (model_variant, seed), df_group in metric_df.groupby(["model_variant", "seed"], dropna=False):
        df_group = df_group.sort_values("height_class")
        label = f"{model_variant} | seed={seed}"
        plt.plot(df_group["height_class"], df_group["Bias"], marker="o", label=label)

    plt.axhline(0, linewidth=1)
    plt.xlabel("Classe de altura")
    plt.ylabel("Bias médio (m)")
    plt.title("Bias por classe de altura — testes focados")
    plt.grid(alpha=0.3)
    plt.legend(fontsize=7, ncol=2)
    plt.tight_layout()

    fig_path = PLOTS_DIR / "bias_por_classe_altura_arquiteturas.png"
    plt.savefig(fig_path, dpi=150, bbox_inches="tight")
    print(f"Figura guardada: {fig_path}")
    plt.show()


def plot_true_vs_pred_for_model(predictions_df, model_variant, seed):
    df = predictions_df[
        (predictions_df["model_variant"] == model_variant)
        & (predictions_df["seed"].astype(str) == str(seed))
    ].copy()

    if df.empty:
        print(f"Sem dados para {model_variant}, seed={seed}")
        return

    metrics = regression_metrics(df["y_true_m"], df["y_pred_m"])

    plt.figure(figsize=(6, 6))
    plt.scatter(df["y_true_m"], df["y_pred_m"], alpha=0.35, s=12)

    mn = min(df["y_true_m"].min(), df["y_pred_m"].min())
    mx = max(df["y_true_m"].max(), df["y_pred_m"].max())
    plt.plot([mn, mx], [mn, mx], linestyle="--", linewidth=1)

    plt.xlabel("Altura real (m)")
    plt.ylabel("Altura prevista (m)")
    plt.title(
        f"{model_variant} | seed={seed}\n"
        f"RMSE={metrics['RMSE']:.3f} m | %RMSE={metrics['%RMSE']:.2f}% | "
        f"R²={metrics['R2']:.3f} | Bias={metrics['Bias']:.3f} m"
    )
    plt.grid(alpha=0.3)
    plt.tight_layout()

    safe_model = re.sub(r"[^A-Za-z0-9_]+", "_", str(model_variant))
    safe_seed = re.sub(r"[^A-Za-z0-9_]+", "_", str(seed))
    fig_path = PLOTS_DIR / f"scatter_true_vs_pred_{safe_model}_seed_{safe_seed}.png"
    plt.savefig(fig_path, dpi=150, bbox_inches="tight")
    print(f"Figura guardada: {fig_path}")
    plt.show()


# ============================================================
# 9. CARREGAMENTO DOS DADOS 11x11
# ============================================================

print("=" * 80)
print(" LATE/EARLY CNN 11x11 — FOCUSED FOLLOW-UP TESTS")
print("=" * 80)

X_all, y_all, regions_all, pids_all, band_names_ref = load_all_regions_tifs(
    RAW_BASE_DIR,
    VALIDATION_DIR,
    target_column=TARGET_COLUMN,
)

print(f"\nX_all: {X_all.shape}")
print(f"y_all: {y_all.shape}")
print(f"Regiões: {sorted(np.unique(regions_all))}")
print(f"N amostras: {len(y_all)}")

CHANNEL_GROUPS = infer_late_fusion_channel_groups(band_names_ref)

height_classes, strata, fixed_bins_m, height_class_labels, cv_strata_source = choose_region_fixed_height_strata_for_cv(
    y_all,
    regions_all,
    n_splits=N_SPLITS,
)

HEIGHT_CLASS_LABEL_MAP = {i: label for i, label in enumerate(height_class_labels)}
print_fixed_height_bins(fixed_bins_m, height_class_labels)
print(f"Origem dos estratos da CV: {cv_strata_source}")

# Guardar distribuição dos dados.
data_distribution_df = pd.DataFrame({
    "region": regions_all,
    "point_id": pids_all,
    "y_cm": y_all,
    "y_m": y_all / TARGET_SCALE,
    "height_class": height_classes,
    "height_class_label": [HEIGHT_CLASS_LABEL_MAP[c] for c in height_classes],
    "stratum": strata,
    "cv_strata_source": cv_strata_source,
})

data_distribution_df.to_csv(
    OUTPUT_BASE_DIR / "dataset_distribution_11x11.csv",
    index=False,
)

print("\n=== Distribuição por região ===")
display(data_distribution_df["region"].value_counts().rename_axis("region").reset_index(name="n"))

print("\n=== Distribuição por classe de altura ===")
display(data_distribution_df["height_class"].value_counts().sort_index().rename_axis("height_class").reset_index(name="n"))

print("\n=== Distribuição região × classe de altura ===")
display(pd.crosstab(data_distribution_df["region"], data_distribution_df["height_class"]))


# ============================================================
# 10. FOLDS CV
# ============================================================

skf = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=SEED,
)

cv_splits = list(skf.split(X_all, strata))

fold_ids = np.full(len(y_all), fill_value=-1, dtype=int)
for fold, (_, val_idx) in enumerate(cv_splits, start=1):
    fold_ids[val_idx] = fold

fold_assignments_df = data_distribution_df.copy()
fold_assignments_df["fold"] = fold_ids
fold_assignments_df["patch_size"] = PATCH_SIZE_TO_USE
fold_assignments_df["patch_area_m"] = PATCH_SIZE_TO_USE * 25

fold_assignments_df.to_csv(
    OUTPUT_BASE_DIR / "fold_assignments_cv_global_misturado_11x11.csv",
    index=False,
)

print("\n=== Distribuição de regiões por fold ===")
display(pd.crosstab(fold_assignments_df["fold"], fold_assignments_df["region"]))

print("\n=== Distribuição de classes de altura por fold ===")
display(pd.crosstab(fold_assignments_df["fold"], fold_assignments_df["height_class"]))


# ============================================================
# 11. DEFINIR TESTES FOCADOS
# ============================================================

# Defaults defensivos, caso a célula de configuração não tenha sido atualizada.
ARCHITECTURE_SEEDS = globals().get("ARCHITECTURE_SEEDS", [42])
RUN_CORE_FOCUSED_TESTS = globals().get("RUN_CORE_FOCUSED_TESTS", True)
RUN_WIDTH_TESTS = globals().get("RUN_WIDTH_TESTS", True)
RUN_WEIGHTING_TESTS = globals().get("RUN_WEIGHTING_TESTS", True)
MAX_ARCHITECTURE_EXPERIMENTS = globals().get("MAX_ARCHITECTURE_EXPERIMENTS", None)

DEFAULT_ARCH_CONFIG = {
    "fusion_type": "late",
    "block_type": "conv",
    "depth": 1,
    "branch_channels": 32,
    "pooling_type": "gap2_center",
    "center_radius": 1,
    "dilation_mode": "last",
    "regressor_type": "current_mlp",
    "dropout": 0.3,
    "se_reduction": 8,
}

experiments = []
seen_experiment_keys = set()


def make_arch_config(**overrides):
    cfg = dict(DEFAULT_ARCH_CONFIG)
    cfg.update(overrides)
    return cfg


def _make_hashable_loss_config(loss_config):
    items = []
    for k, v in sorted(dict(loss_config).items()):
        if isinstance(v, list):
            v = tuple(v)
        if isinstance(v, np.ndarray):
            v = tuple(v.tolist())
        items.append((k, v))
    return tuple(items)


def _experiment_key(cfg, groups_label, sampling_mode, loss_config, sampler_label):
    relevant = (
        cfg.get("fusion_type"), cfg.get("block_type"), int(cfg.get("depth")),
        int(cfg.get("branch_channels")), cfg.get("pooling_type"),
        int(cfg.get("center_radius")), cfg.get("dilation_mode"),
        cfg.get("regressor_type"), groups_label, sampling_mode, sampler_label,
        _make_hashable_loss_config(loss_config),
    )
    return relevant


def add_arch_experiment(
    tag,
    ablation_type,
    overrides=None,
    groups_to_keep="all",
    seeds=None,
    sampling_mode="standard",
    sampler_classes_all=None,
    sampler_label="none",
    loss_config=None,
):
    if overrides is None:
        overrides = {}
    if seeds is None:
        seeds = ARCHITECTURE_SEEDS
    if loss_config is None:
        loss_config = {"loss_mode": "standard"}

    cfg = make_arch_config(**overrides)
    channel_groups = filter_channel_groups(CHANNEL_GROUPS, groups_to_keep=groups_to_keep)
    groups_label = describe_channel_groups(channel_groups)
    key = _experiment_key(cfg, groups_label, sampling_mode, loss_config, sampler_label)

    if key in seen_experiment_keys:
        return
    seen_experiment_keys.add(key)

    model_variant = f"{tag}_{architecture_label(cfg)}"
    model_variant = re.sub(r"[^A-Za-z0-9_]+", "_", model_variant)

    experiments.append({
        "model_variant": model_variant,
        "ablation_type": ablation_type,
        "sampling_mode": sampling_mode,
        "sampler_label": sampler_label,
        "seeds": list(seeds),
        "sampler_classes_all": sampler_classes_all,
        "loss_config": dict(loss_config),
        "channel_groups": channel_groups,
        "architecture_config": cfg,
    })


# Classes combinadas para sampler região × altura.
region_height_classes = np.array([
    f"{region}__h{height_class}"
    for region, height_class in zip(regions_all, height_classes)
])

# ------------------------------------------------------------
# 11.1 Núcleo arquitetural: juntar os melhores sinais da ablação anterior
# ------------------------------------------------------------
if RUN_CORE_FOCUSED_TESTS:
    # Melhor arquitetura anterior da nova ablação: depth1 + late + gap2_center + current MLP.
    add_arch_experiment(
        "A1_best_previous_depth1",
        "focused_architecture",
        {"fusion_type": "late", "depth": 1, "pooling_type": "gap2_center", "regressor_type": "current_mlp"},
    )

    # Testar se o regressor linear melhora a versão depth1.
    add_arch_experiment(
        "A2_depth1_linear",
        "focused_architecture",
        {"fusion_type": "late", "depth": 1, "pooling_type": "gap2_center", "regressor_type": "linear"},
    )

    # Testar se o center_only melhora quando combinado com depth1.
    add_arch_experiment(
        "A3_depth1_center_only_current",
        "focused_architecture",
        {"fusion_type": "late", "depth": 1, "pooling_type": "center_only", "regressor_type": "current_mlp"},
    )

    # Combinação mais simples sugerida pelos resultados: depth1 + center_only + linear.
    add_arch_experiment(
        "A4_depth1_center_only_linear",
        "focused_architecture",
        {"fusion_type": "late", "depth": 1, "pooling_type": "center_only", "regressor_type": "linear"},
    )

    # Early fusion também com depth1, para comparação justa com late fusion.
    add_arch_experiment(
        "A5_early_depth1_current",
        "focused_architecture",
        {"fusion_type": "early", "depth": 1, "pooling_type": "gap2_center", "regressor_type": "current_mlp"},
    )

    add_arch_experiment(
        "A6_early_depth1_linear",
        "focused_architecture",
        {"fusion_type": "early", "depth": 1, "pooling_type": "gap2_center", "regressor_type": "linear"},
    )

    add_arch_experiment(
        "A7_early_depth1_center_only_linear",
        "focused_architecture",
        {"fusion_type": "early", "depth": 1, "pooling_type": "center_only", "regressor_type": "linear"},
    )

# ------------------------------------------------------------
# 11.2 Largura/capacidade: testar se ch32 está demasiado grande ou pequeno
# ------------------------------------------------------------
if RUN_WIDTH_TESTS:
    add_arch_experiment(
        "B1_late_depth1_center_linear_ch16",
        "width_ablation",
        {"fusion_type": "late", "depth": 1, "branch_channels": 16, "pooling_type": "center_only", "regressor_type": "linear"},
    )

    add_arch_experiment(
        "B2_late_depth1_center_linear_ch64",
        "width_ablation",
        {"fusion_type": "late", "depth": 1, "branch_channels": 64, "pooling_type": "center_only", "regressor_type": "linear"},
    )

# ------------------------------------------------------------
# 11.3 Loss/sampler: tentar reduzir bias por altura
# ------------------------------------------------------------
if RUN_WEIGHTING_TESTS:
    base_arch = {"fusion_type": "late", "depth": 1, "pooling_type": "gap2_center", "regressor_type": "current_mlp"}

    add_arch_experiment(
        "C1_depth1_sampler_height",
        "height_balance",
        base_arch,
        sampling_mode="weighted_sampler",
        sampler_classes_all=height_classes,
        sampler_label="height_class",
        loss_config={"loss_mode": "standard"},
    )

    add_arch_experiment(
        "C2_depth1_sampler_region_height",
        "height_region_balance",
        base_arch,
        sampling_mode="weighted_sampler",
        sampler_classes_all=region_height_classes,
        sampler_label="region_height_class",
        loss_config={"loss_mode": "standard"},
    )

    add_arch_experiment(
        "C3_depth1_class_weighted_mse",
        "height_balance_loss",
        base_arch,
        loss_config={
            "loss_mode": "class_weighted",
            "base_loss": "mse",
            "class_bins_m": tuple(FIXED_HEIGHT_BINS_M),
            "class_weight_power": 1.0,
        },
    )

    add_arch_experiment(
        "C4_depth1_class_weighted_smoothl1",
        "height_balance_loss",
        base_arch,
        loss_config={
            "loss_mode": "class_weighted_smooth_l1",
            "beta": 1.0,
            "class_bins_m": tuple(FIXED_HEIGHT_BINS_M),
            "class_weight_power": 1.0,
        },
    )

    # Mesmo teste, mas aplicado à arquitetura mais simples.
    add_arch_experiment(
        "C5_center_linear_class_weighted_smoothl1",
        "height_balance_loss",
        {"fusion_type": "late", "depth": 1, "pooling_type": "center_only", "regressor_type": "linear"},
        loss_config={
            "loss_mode": "class_weighted_smooth_l1",
            "beta": 1.0,
            "class_bins_m": tuple(FIXED_HEIGHT_BINS_M),
            "class_weight_power": 1.0,
        },
    )

if MAX_ARCHITECTURE_EXPERIMENTS is not None:
    experiments = experiments[:int(MAX_ARCHITECTURE_EXPERIMENTS)]

# Guardar configuração das experiências.
experiments_config_df = pd.DataFrame([
    {
        "model_variant": exp["model_variant"],
        "ablation_type": exp["ablation_type"],
        "sampling_mode": exp["sampling_mode"],
        "sampler_label": exp.get("sampler_label", "none"),
        "seeds": ",".join(map(str, exp["seeds"])),
        "active_groups": describe_channel_groups(exp["channel_groups"]),
        **exp["architecture_config"],
        **exp["loss_config"],
    }
    for exp in experiments
])
experiments_config_df.to_csv(OUTPUT_BASE_DIR / "focused_experiments_config.csv", index=False)

print("\nExperiências focadas a correr:")
display(experiments_config_df)
print(f"\nTotal de experiências: {len(experiments)}")
print(f"Seeds por experiência: {ARCHITECTURE_SEEDS}")
print(f"Treinos totais previstos: {len(experiments) * len(ARCHITECTURE_SEEDS) * N_SPLITS}")

if len(experiments) == 0:
    raise ValueError("Nenhuma experiência foi definida. Confirma os flags RUN_*.")

train_transform = get_train_transform(AUGMENTATION_MODE)


# ============================================================
# 12. TREINO
# ============================================================

all_results = []
all_predictions = []
all_histories = []

for exp in experiments:
    model_variant = exp["model_variant"]
    sampling_mode = exp["sampling_mode"]
    seeds = exp["seeds"]
    sampler_classes_all = exp.get("sampler_classes_all")
    loss_config = exp.get("loss_config", {"loss_mode": "standard"})
    architecture_config = exp.get("architecture_config", DEFAULT_ARCH_CONFIG)
    CURRENT_SAMPLER_LABEL = exp.get("sampler_label", "none")

    for seed in seeds:
        for fold, (train_idx, val_idx) in enumerate(cv_splits, start=1):
            results_df, history_df, predictions_df = run_fold_for_variant(
                X_all=X_all,
                y_all=y_all,
                regions_all=regions_all,
                pids_all=pids_all,
                height_classes_all=height_classes,
                sampler_classes_all=sampler_classes_all,
                train_idx=train_idx,
                val_idx=val_idx,
                fold=fold,
                model_variant=model_variant,
                sampling_mode=sampling_mode,
                seed=seed,
                train_transform=train_transform,
                loss_config=loss_config,
                channel_groups=exp.get("channel_groups", CHANNEL_GROUPS),
                architecture_config=architecture_config,
                ablation_type=exp.get("ablation_type", "architecture"),
            )

            all_results.append(results_df)
            all_predictions.append(predictions_df)
            all_histories.append(history_df)

individual_results_df = pd.concat(all_results, ignore_index=True)
individual_predictions_df = pd.concat(all_predictions, ignore_index=True)
all_histories_df = pd.concat(all_histories, ignore_index=True)

individual_results_df.to_csv(
    OUTPUT_BASE_DIR / "resultados_individuais_por_seed_fold.csv",
    index=False,
)

individual_predictions_df.to_csv(
    OUTPUT_BASE_DIR / "predicoes_individuais_por_seed.csv",
    index=False,
)

all_histories_df.to_csv(
    OUTPUT_BASE_DIR / "historicos_treino_por_seed_fold.csv",
    index=False,
)

print("\n=== Resultados individuais por seed/fold ===")
display(individual_results_df)


# ============================================================
# 13. ENSEMBLE DAS SEEDS, SE HOUVER MAIS DO QUE UMA SEED
# ============================================================

ensemble_predictions_list = []

for exp in experiments:
    if len(exp["seeds"]) <= 1:
        continue

    ens_df = build_ensemble_predictions(
        individual_predictions_df,
        model_variant=exp["model_variant"],
        sampling_mode=exp["sampling_mode"],
        seeds=exp["seeds"],
    )

    if not ens_df.empty:
        ensemble_predictions_list.append(ens_df)

if ensemble_predictions_list:
    ensemble_predictions_df = pd.concat(ensemble_predictions_list, ignore_index=True)
else:
    ensemble_predictions_df = pd.DataFrame()

ensemble_predictions_df.to_csv(
    OUTPUT_BASE_DIR / "predicoes_ensemble.csv",
    index=False,
)

all_predictions_for_comparison = pd.concat(
    [individual_predictions_df, ensemble_predictions_df],
    ignore_index=True,
)

all_predictions_for_comparison.to_csv(
    OUTPUT_BASE_DIR / "TODAS_predicoes_comparacao.csv",
    index=False,
)


# ============================================================
# 14. MÉTRICAS FINAIS
# ============================================================

fold_metrics_df = metrics_by_fold_from_predictions(all_predictions_for_comparison)
fold_metrics_df.to_csv(
    OUTPUT_BASE_DIR / "metricas_por_fold_todos_modelos.csv",
    index=False,
)

summary_metrics_df = summarize_fold_metrics(fold_metrics_df)
summary_metrics_df.to_csv(
    OUTPUT_BASE_DIR / "comparacao_modelos_media_std_folds.csv",
    index=False,
)

region_group_cols = [
    "ablation_type", "architecture", "fusion_type", "block_type", "depth",
    "pooling_type", "regressor_type", "branch_channels", "center_radius",
    "center_window", "dilation_mode", "active_groups", "n_active_channels",
    "n_params", "n_trainable_params", "model_variant", "sampling_mode",
    "loss_mode", "augmentation", "seed", "region",
]
region_group_cols = [c for c in region_group_cols if c in all_predictions_for_comparison.columns]
metrics_by_region_df = grouped_regression_metrics(
    all_predictions_for_comparison,
    group_cols=region_group_cols,
)
metrics_by_region_df.to_csv(
    OUTPUT_BASE_DIR / "metricas_por_regiao_pooled.csv",
    index=False,
)

height_group_cols = [
    "ablation_type", "architecture", "fusion_type", "block_type", "depth",
    "pooling_type", "regressor_type", "branch_channels", "center_radius",
    "center_window", "dilation_mode", "active_groups", "n_active_channels",
    "n_params", "n_trainable_params", "model_variant", "sampling_mode",
    "loss_mode", "augmentation", "seed", "height_class", "height_class_label",
]
height_group_cols = [c for c in height_group_cols if c in all_predictions_for_comparison.columns]
metrics_by_height_class_df = grouped_regression_metrics(
    all_predictions_for_comparison,
    group_cols=height_group_cols,
)
metrics_by_height_class_df.to_csv(
    OUTPUT_BASE_DIR / "metricas_por_classe_altura_pooled.csv",
    index=False,
)

# Resumo só das médias por arquitetura.
summary_architecture_df = (
    fold_metrics_df
    .groupby([
        "ablation_type", "model_variant", "architecture", "fusion_type", "block_type",
        "depth", "pooling_type", "regressor_type", "branch_channels", "center_radius",
        "n_params", "n_trainable_params", "seed",
    ], dropna=False)[["RMSE", "%RMSE", "MAE", "R2", "Bias"]]
    .agg(["mean", "std"])
    .reset_index()
)
summary_architecture_df.to_csv(
    OUTPUT_BASE_DIR / "RESUMO_arquiteturas_media_std_folds.csv",
    index=False,
)

print("\n=== Métricas por fold — todos os modelos ===")
display(fold_metrics_df)

print("\n=== Comparação final — média ± std das folds ===")
display(summary_metrics_df)

print("\n=== Resumo arquitetural — média ± std das folds ===")
display(summary_architecture_df)

print("\n=== Métricas por região — pooled predictions ===")
display(metrics_by_region_df)

print("\n=== Métricas por classe de altura — pooled predictions ===")
display(metrics_by_height_class_df)


# ============================================================
# 15. GRÁFICOS FINAIS
# ============================================================

plot_metric_comparison(fold_metrics_df, metric="%RMSE")
plot_metric_comparison(fold_metrics_df, metric="RMSE")
plot_metric_comparison(fold_metrics_df, metric="MAE")
plot_bias_by_height_class(all_predictions_for_comparison)

# Scatter dos melhores modelos por %RMSE médio, para não gerar dezenas de figuras.
best_models = (
    fold_metrics_df
    .groupby(["model_variant", "seed"], dropna=False)[["RMSE", "%RMSE", "MAE", "R2", "Bias"]]
    .mean()
    .reset_index()
    .sort_values("%RMSE")
    .head(5)
)

print("\n=== Top 5 modelos por %RMSE médio ===")
display(best_models)

for _, row in best_models.iterrows():
    plot_true_vs_pred_for_model(
        all_predictions_for_comparison,
        model_variant=row["model_variant"],
        seed=row["seed"],
    )

plot_error_distribution_by_height(
    all_predictions_for_comparison,
    top_models_df=best_models.head(3),
    metric_name="abs_error_m",
)

# Resumo final ordenado explicitamente por %RMSE.
summary_ranked_df = (
    fold_metrics_df
    .groupby([
        "ablation_type", "model_variant", "fusion_type", "block_type", "depth",
        "pooling_type", "regressor_type", "branch_channels", "sampling_mode",
        "loss_mode", "seed",
    ], dropna=False)[["RMSE", "%RMSE", "MAE", "R2", "Bias"]]
    .agg(["mean", "std"])
    .reset_index()
)
summary_ranked_df = flatten_summary_columns(summary_ranked_df)
summary_ranked_df = summary_ranked_df.sort_values("%RMSE_mean")
summary_ranked_df.to_csv(OUTPUT_BASE_DIR / "RANKING_FINAL_ordenado_por_pctRMSE.csv", index=False)

print("\n=== Ranking final ordenado por %RMSE ===")
display(summary_ranked_df)

print(f"\nResultados guardados em: {OUTPUT_BASE_DIR}")
```
